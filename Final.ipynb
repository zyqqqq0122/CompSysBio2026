# Tumor Prediction — TCGA-LUAD Multi-Omics

This notebook trains and evaluates **tumour vs. normal classifiers** using three omics modalities
individually and in pairwise early-fusion combinations.

| Layer | Samples | Features |
|-------|---------|----------|
| RNASeq | 447 | 14 434 genes |
| DNAm | 420 | 384 629 CpG sites |
| CNV | 424 | 14 434 genes |

**Pipeline per experiment**
1. Variance filter — keep top *N* most variable features per modality
2. StandardScaler → fit classifier → 5-fold stratified cross-validation
3. Report AUC-ROC, Balanced Accuracy, F1 (macro), Accuracy

**Models**
- Logistic Regression (L2, PCA-compressed, class-weighted)
- Random Forest (200 trees, PCA-compressed, class-weighted)
- Linear SVM (Platt-calibrated probabilities, PCA-compressed, class-weighted)
- Elastic Net (L1+L2 logistic regression, raw features, class-weighted)
- XGBoost (gradient boosting, raw features)

```python
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.base import clone
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.pipeline import Pipeline
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import (
    roc_auc_score, balanced_accuracy_score,
    f1_score, accuracy_score, roc_curve, auc
)
from xgboost import XGBClassifier

sns.set_theme(style='whitegrid', palette='muted')
plt.rcParams['figure.dpi'] = 120

print('Libraries loaded.')
```

```python
# Sample-ID format audit — run before harmonisation to document all formats
# and confirm there are no unresolvable ID mismatches between datasets.
print('=== Sample ID format audit (raw, before harmonisation) ===\n')

def _id_format_summary(name, idx):
    lengths = pd.Series(idx.str.len())
    has_dot  = idx.str.contains('.', regex=False).sum()
    has_dash = idx.str.contains('-').sum()
    print(f'{name}  ({len(idx)} samples)')
    print(f'  ID lengths : {sorted(lengths.unique().tolist())}')
    print(f'  Dot sep    : {has_dot}   Dash sep : {has_dash}')
    for l in sorted(lengths.unique().tolist()):
        ex = idx[lengths == l][:2].tolist()
        print(f'  len={l:2d} ({(lengths == l).sum():3d} samples)  e.g. {ex}')
    print()

_id_format_summary('metadata patient_id', pd.Index(meta['patient_id']))
_id_format_summary('RNASeq',              rnaseq.index)
_id_format_summary('DNAm',               dnam.index)
_id_format_summary('CNV',                cnv.index)

# Verify that ALL dataset IDs resolve to a known metadata sample after the new
# normalisation (same field-clip logic as cell-harmonise).
def _norm_preview(name, idx):
    """Preview normalisation: dot→dash + dataset-specific field clip."""
    idx = idx.str.replace('.', '-', regex=False)
    if name in ('RNASeq', 'DNAm'):
        idx = idx.str.split('-').str[:4].str.join('-')
        if name == 'DNAm':
            parts = idx.str.split('-')
            idx = pd.Index([
                '-'.join(p[:3]) if p[3].startswith('0') else '-'.join(p[:4])
                for p in parts
            ])
    # CNV: dot→dash only (IDs already in 3-field tumour or 4-field normal format)
    return idx

meta_set = set(meta['patient_id'].unique())
for name, idx in [('RNASeq', rnaseq.index), ('DNAm', dnam.index), ('CNV', cnv.index)]:
    normed    = _norm_preview(name, idx)
    n_miss    = (~normed.isin(meta_set)).sum()
    n_dup_raw = idx.duplicated().sum()
    # Patients with entries of more than one normalised form (3-field tumour + 4-field normal)
    pid3   = normed.str.split('-').str[:3].str.join('-')
    n_both = sum(
        1 for pid in pid3.unique()
        if len(set(normed[pid3 == pid])) > 1
    )
    print(f'{name:8s}  unresolved after norm={n_miss}  '
          f'raw dupes={n_dup_raw}  '
          f'patients with tumor+normal entries={n_both}')
```

```python
# Normalise TCGA sample IDs to match the patient_id format used in metadata.
#
# Metadata contains two ID formats:
#   3-field  TCGA-XX-XXXX        → patient-level tumour entry   (label = 1)
#   4-field  TCGA-XX-XXXX-11A   → sample-level normal-tissue entry (label = 0)
#
# Raw ID formats across the three source files and how they are normalised:
#
#   File    Raw example                        Normalisation              Result
#   ------  ---------------------------------  -------------------------  ------------------
#   RNASeq  TCGA.69.7765                       replace . → -             TCGA-69-7765  (3-field)
#   RNASeq  TCGA-44-2655-11A-01R-1758-07      clip to 4 fields          TCGA-44-2655-11A (4-field)
#   DNAm    TCGA.69.8255.01A.11D.2285.05      replace . → -, clip×4,   TCGA-69-8255  (3-field)
#                                               4th field '01x' → clip×3
#   DNAm    TCGA.50.5932.11A.01D.1756.05      replace . → -, clip×4    TCGA-50-5932-11A (4-field)
#   CNV     TCGA.69.7765                       replace . → -             TCGA-69-7765  (3-field)
#   CNV     TCGA.55.6983.11A                   replace . → -             TCGA-55-6983-11A (4-field)
#
# Keeping sample-level IDs means a patient with both a tumour sample (TCGA-XX-XXXX)
# and a normal-tissue sample (TCGA-XX-XXXX-11A) has TWO rows with DISTINCT IDs and
# CORRECT labels — no deduplication, no mislabelling.

# ── RNASeq ───────────────────────────────────────────────────────────────────
# Tumour IDs are already 3-field (TCGA.XX.XXXX); normal IDs are 7-field full
# barcodes (TCGA-XX-XXXX-11A-01R-YYYY-07).  Clip to 4 fields after dot→dash.
rnaseq.index = (rnaseq.index
                .str.replace('.', '-', regex=False)
                .str.split('-').str[:4].str.join('-'))

# ── DNAm ─────────────────────────────────────────────────────────────────────
# All DNAm IDs are 7-field full barcodes.  After dot→dash and 4-field clip:
#   tumour barcodes → TCGA-XX-XXXX-01A  (4th field starts with '0')
#   normal barcodes → TCGA-XX-XXXX-11A  (4th field starts with '1')
# Tumour entries need a further clip to 3 fields to match the 3-field
# patient-level IDs used for tumour entries in metadata.
dnam.index = dnam.index.str.replace('.', '-', regex=False)
dnam.index = dnam.index.str.split('-').str[:4].str.join('-')
_p = dnam.index.str.split('-')
dnam.index = pd.Index([
    '-'.join(p[:3]) if p[3].startswith('0') else '-'.join(p[:4])
    for p in _p
])

# ── CNV ──────────────────────────────────────────────────────────────────────
# CNV IDs are already in meta format (3-field tumour or 4-field normal).
# Only dot→dash normalisation required.
cnv.index = cnv.index.str.replace('.', '-', regex=False)

# ── Label map ────────────────────────────────────────────────────────────────
# Metadata already uses dashes.  Build an exact-match label map.
# Dedup handles the few cases where the same patient_id appears twice in meta.
label_map = pd.Series(meta['label'].values, index=meta['patient_id'])
label_map = label_map[~label_map.index.duplicated(keep='first')]

# ── Coverage check ───────────────────────────────────────────────────────────
print(f'Metadata label map: {len(label_map)} unique IDs')
print(f'  label=0 (Normal): {(label_map == 0).sum()}')
print(f'  label=1 (Tumour): {(label_map == 1).sum()}\n')

for name, df in [('RNASeq', rnaseq), ('DNAm', dnam), ('CNV', cnv)]:
    labels      = df.index.map(label_map)
    n_matched   = labels.notna().sum()
    n_unmatched = labels.isna().sum()
    n_normal    = int((labels == 0).sum())
    n_tumour    = int((labels == 1).sum())
    suffix = f'  WARNING: {n_unmatched} unmatched IDs' if n_unmatched else ''
    print(f'{name:8s}  {len(df):4d} samples  →  {n_matched} matched  '
          f'(Normal={n_normal}  Tumour={n_tumour}){suffix}')
```

## 2. Preprocessing & Feature Selection

The pipeline applies **two sequential dimensionality-reduction steps**:

1. **Variance filter** — retain the top *N* most variable features per modality.
   Fast unsupervised pre-filtering to a tractable range before PCA.
   For DNAm this is especially important: the 384 629 CpG sites are massively
   correlated because many belong to the same gene or regulatory region,
   so even the top-2000 subset will contain strong co-linear structure.

2. **PCA (Principal Component Analysis)** — applied *inside the sklearn pipeline*,
   fit only on training data in each CV fold to prevent leakage.
   PCA decorrelates the filtered features and compresses them into *K* orthogonal
   components that capture the dominant axes of variation.

**Pairwise early fusion** applies modality-specific scaling + PCA to each block
independently (`ColumnTransformer`), then concatenates the *K* components from
both modalities. This prevents one modality from dominating the joint feature space.

```python
def select_top_var(df, n):
    return df[df.var(axis=0).nlargest(n).index]


def prepare_single(df, label_map, n_features):
    """
    Variance filter → label lookup via patient ID → (X, y).
    Samples whose patient ID is absent from label_map are dropped.
    """
    df_sel = select_top_var(df, n_features)
    labels = df_sel.index.map(label_map)
    valid  = ~labels.isna()
    X = df_sel.loc[valid].values.astype(float)
    y = labels[valid].astype(int).values
    return X, y


def prepare_pair(df1, df2, label_map, n_features, prefix1, prefix2):
    """
    Concatenate top features from two modalities on their shared patient IDs.
    Label is looked up from metadata via the (already-normalised) 12-char patient ID.
    Samples without a metadata match are excluded.
    """
    sel1 = select_top_var(df1, n_features).add_prefix(prefix1 + '_')
    sel2 = select_top_var(df2, n_features).add_prefix(prefix2 + '_')
    shared = sel1.index.intersection(sel2.index)
    X      = np.hstack([sel1.loc[shared].values,
                        sel2.loc[shared].values]).astype(float)
    labels = pd.Series(shared.map(label_map), index=shared)
    valid  = ~labels.isna()
    X = X[valid.values]
    y = labels[valid].astype(int).values
    return X, y


print('Preprocessing helpers defined.')
```

## 3. Classifiers & Evaluation Framework

**PCA policy (applies to ALL models):**
- **RNASeq** and **DNAm**: always compressed via `StandardScaler → PCA(K)` before the classifier.
- **CNV**: always uses raw variance-filtered features (`StandardScaler` only — no PCA).

**Single-omics pipeline per fold:**
> Variance filter → StandardScaler [→ PCA(*K*) if RNA/DNAm] → Classifier

**Pairwise pipeline per fold:**
> [mod 1: Scaler → PCA(*K*) if RNA/DNAm] ⊕ [mod 2: Scaler → PCA(*K*) if RNA/DNAm] → Concat → Classifier
> Each modality block processed independently via `ColumnTransformer`.

All class-weighted models use `class_weight='balanced'` to handle tumour/normal imbalance.

| Model | Regularisation | PCA (RNA/DNAm) | PCA (CNV) |
|-------|---------------|:--------------:|:---------:|
| Logistic Regression | L2, C = 1.0 | ✓ | ✗ |
| Random Forest | 200 trees | ✓ | ✗ |
| Linear SVM | C = 1.0, Platt-scaled | ✓ | ✗ |
| Elastic Net | L1+L2, l1_ratio = 0.5, C = 1.0 | ✓ | ✗ |
| XGBoost | 100 trees, lr = 0.05, depth = 3 | ✓ | ✗ |

**Active models** are controlled by `MODELS_TO_RUN` in the config cell.

**Evaluation**: 5-fold stratified CV. All preprocessing steps are re-fit on training data each fold — no leakage.

```python
# PCA diagnostic: explained variance per modality
# Fitted on all labelled samples to guide the N_PCA choice.
# In the actual classification pipeline, PCA is re-fit on each training fold only.
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, (name, df, col) in zip(axes, [('RNASeq', rnaseq, PALETTE['RNASeq']),
                                        ('DNAm',   dnam,   PALETTE['DNAm']),
                                        ('CNV',    cnv,    PALETTE['CNV'])]):
    labels = df.index.map(label_map)
    df_sel = select_top_var(df, N_SINGLE)
    valid  = ~labels.isna()
    X_sub  = df_sel.loc[valid].values.astype(float)

    n_show = min(N_PCA * 2, 100)
    n_comp = min(n_show, X_sub.shape[0] - 1, X_sub.shape[1])
    pca_diag = PCA(n_components=n_comp, random_state=SEED)
    pca_diag.fit(StandardScaler().fit_transform(X_sub))

    evr    = pca_diag.explained_variance_ratio_ * 100
    cumvar = np.cumsum(evr)
    xs     = range(1, len(evr) + 1)

    ax.bar(xs, evr, color=col, alpha=0.6, label='Per-PC variance')
    ax2 = ax.twinx()
    ax2.plot(xs, cumvar, 'r-o', markersize=3, lw=1.5, label='Cumulative %')
    ax2.set_ylim(0, 105)
    ax2.set_ylabel('Cumulative variance (%)', fontsize=8)

    ax.axvline(N_PCA, ls='--', color='black', lw=1.5, label=f'N_PCA = {N_PCA}')
    ax2.axhline(95, ls=':', color='grey', lw=1, alpha=0.8)

    n_for_95   = int(np.argmax(cumvar >= 95)) + 1
    cum_at_cut = cumvar[min(N_PCA - 1, len(cumvar) - 1)]
    ax.text(0.97, 0.55,
            f'PCs for 95% var: {n_for_95}\nVar at PC {N_PCA}: {cum_at_cut:.1f}%',
            transform=ax.transAxes, ha='right', va='bottom', fontsize=8,
            bbox=dict(boxstyle='round', facecolor='white', alpha=0.85))

    ax.set_title(f'{name}  (top {N_SINGLE} features → PCA, n={valid.sum()})', fontsize=10)
    ax.set_xlabel('Principal Component')
    ax.set_ylabel('Variance explained (%)')
    lines1, labs1 = ax.get_legend_handles_labels()
    lines2, labs2 = ax2.get_legend_handles_labels()
    ax.legend(lines1 + lines2, labs1 + labs2, fontsize=7, loc='upper right')

plt.suptitle('PCA scree plots per modality\n'
             '(diagnostic fit on all labelled samples; actual CV fits only on training folds)',
             fontsize=12)
plt.tight_layout()
plt.show()
```

```python
def make_single_models(use_pca=True):
    """
    Build all five classifiers with a uniform PCA policy.
    use_pca=True  → StandardScaler → PCA(N_PCA) → Classifier  (RNA and DNAm)
    use_pca=False → StandardScaler → Classifier               (CNV)
    Only models listed in MODELS_TO_RUN are returned.
    """
    def build(clf):
        steps = [('scaler', StandardScaler())]
        if use_pca:
            steps.append(('pca', PCA(n_components=N_PCA, random_state=SEED)))
        steps.append(('clf', clf))
        return Pipeline(steps)

    all_models = {
        'Logistic Regression': build(LogisticRegression(
            C=1.0, max_iter=2000, class_weight='balanced',
            solver='saga', random_state=SEED)),
        'Random Forest': build(RandomForestClassifier(
            n_estimators=200, max_features='sqrt',
            class_weight='balanced', random_state=SEED, n_jobs=-1)),
        'Linear SVM': build(SVC(
            kernel='linear', C=1.0, probability=True,
            class_weight='balanced', random_state=SEED)),
        'Elastic Net': build(LogisticRegression(
            penalty='elasticnet', solver='saga', l1_ratio=0.5, C=1.0,
            max_iter=5000, class_weight='balanced', random_state=SEED)),
        'XGBoost': build(XGBClassifier(
            n_estimators=100, learning_rate=0.05, max_depth=3,
            eval_metric='logloss', random_state=SEED)),
    }
    return {k: v for k, v in all_models.items() if k in MODELS_TO_RUN}


def make_pair_models(n_per_modality=N_PAIR, pca_mod1=True, pca_mod2=True):
    """
    All classifiers use a per-modality ColumnTransformer so PCA is applied
    independently to each block.
    pca_mod1 / pca_mod2 control whether each modality block is PCA-compressed.
    Only models listed in MODELS_TO_RUN are returned.
    """
    def mod_pipe(use_pca):
        steps = [('scaler', StandardScaler())]
        if use_pca:
            steps.append(('pca', PCA(n_components=N_PCA, random_state=SEED)))
        return Pipeline(steps)

    transform = ColumnTransformer([
        ('mod1', mod_pipe(pca_mod1), slice(0, n_per_modality)),
        ('mod2', mod_pipe(pca_mod2), slice(n_per_modality, 2 * n_per_modality)),
    ])

    classifiers = {
        'Logistic Regression': LogisticRegression(
            C=1.0, max_iter=2000, class_weight='balanced',
            solver='saga', random_state=SEED),
        'Random Forest': RandomForestClassifier(
            n_estimators=200, max_features='sqrt',
            class_weight='balanced', random_state=SEED, n_jobs=-1),
        'Linear SVM': SVC(
            kernel='linear', C=1.0, probability=True,
            class_weight='balanced', random_state=SEED),
        'Elastic Net': LogisticRegression(
            penalty='elasticnet', solver='saga', l1_ratio=0.5, C=1.0,
            max_iter=5000, class_weight='balanced', random_state=SEED),
        'XGBoost': XGBClassifier(
            n_estimators=100, learning_rate=0.05, max_depth=3,
            eval_metric='logloss', random_state=SEED),
    }

    return {
        name: Pipeline([('transform', clone(transform)), ('clf', clf)])
        for name, clf in classifiers.items()
        if name in MODELS_TO_RUN
    }


# Derived from config — used by downstream plotting cells
models_list = list(MODELS_TO_RUN)


def evaluate_dataset(X, y, model_factory, n_splits=CV_FOLDS):
    """
    Stratified K-fold CV for all models from model_factory().
    Automatically reduces n_splits if the minority class is too small.
    All preprocessing steps are re-fit on each training fold only — no data leakage.
    """
    counts = np.bincount(y)
    min_class = counts.min()

    effective_splits = min(n_splits, min_class)
    if effective_splits < 2:
        print(f'    WARNING: only {min_class} minority-class sample(s) — '
              f'cannot run CV, skipping this dataset.')
        nan_metrics = {k: np.nan for k in
                       ['AUC-ROC', 'AUC std', 'Balanced Acc', 'F1 (macro)', 'Accuracy']}
        dummy = {'pool_y_true': y,
                 'pool_y_score': np.full_like(y, 0.5, dtype=float)}
        return {m: {**nan_metrics, **dummy} for m in model_factory().keys()}

    if effective_splits < n_splits:
        print(f'    Note: using {effective_splits}-fold CV '
              f'(minority class has only {min_class} samples)')

    skf = StratifiedKFold(n_splits=effective_splits, shuffle=True, random_state=SEED)
    results = {}

    for mname, pipeline in model_factory().items():
        fold_aucs, fold_baccs, fold_f1s, fold_accs = [], [], [], []
        pool_y_true, pool_y_score = [], []

        for train_idx, test_idx in skf.split(X, y):
            m = clone(pipeline)
            m.fit(X[train_idx], y[train_idx])
            y_pred  = m.predict(X[test_idx])
            y_score = m.predict_proba(X[test_idx])[:, 1]

            fold_aucs.append(roc_auc_score(y[test_idx], y_score))
            fold_baccs.append(balanced_accuracy_score(y[test_idx], y_pred))
            fold_f1s.append(f1_score(y[test_idx], y_pred, average='macro'))
            fold_accs.append(accuracy_score(y[test_idx], y_pred))
            pool_y_true.extend(y[test_idx].tolist())
            pool_y_score.extend(y_score.tolist())

        results[mname] = {
            'AUC-ROC':      round(np.mean(fold_aucs),  4),
            'AUC std':      round(np.std(fold_aucs),   4),
            'Balanced Acc': round(np.mean(fold_baccs), 4),
            'F1 (macro)':   round(np.mean(fold_f1s),   4),
            'Accuracy':     round(np.mean(fold_accs),  4),
            'pool_y_true':  np.array(pool_y_true),
            'pool_y_score': np.array(pool_y_score),
        }

    return results


print('Model factories and evaluation function defined.')
print(f'  Active models: {models_list}')
print(f'  PCA (RNA/DNAm): StandardScaler → PCA({N_PCA}) → Classifier')
print(f'  No PCA (CNV):   StandardScaler → Classifier')
print(f'  Pairwise:       ColumnTransformer([mod1 pipe, mod2 pipe]) → Concat → Classifier')
```

## 4. Single-Omics Classification

Each modality is evaluated independently.
Pipeline per fold: variance filter (top **2 000** features) → StandardScaler → PCA (**50** components) → Classifier.

## 7. DIABLO — Supervised Multi-Omics Integration (all 3 layers)

[DIABLO](https://mixomics.org/methods/diablo/) (`block.splsda` in the `mixOmics` R package)
is a supervised multi-block sparse PLS-DA method. Unlike the pairwise early-fusion pipelines
above, DIABLO integrates **all three omics layers simultaneously** by maximising the
covariance between blocks while retaining discriminant power for the class labels.

Key differences from the sklearn pipeline:

- **No PCA pre-compression**: DIABLO performs its own sparse latent-factor decomposition.
- **Cross-block covariance**: a design matrix controls how strongly blocks are allowed to
  co-vary — off-diagonal entries of 0.1 encourage information sharing without forcing
  identical projections.
- **Feature selection built-in**: `keepX` controls how many features are selected per block
  per component; here 50 features per block.

Performance is estimated by 3-fold × 10-repeat CV via `perf()` — 3 folds because
only 3 Normal samples remain in the shared three-modality intersection (consistent with
how the sklearn pipeline auto-reduces folds for DNAm+CNV).
DIABLO is called from Python via **rpy2** — no PyTorch required.

```python
# ── 7.0  Optional: install mixOmics (run ONCE per environment) ───────────────
# Skip this cell if cell 7.1 prints "mixOmics X.Y loaded."
# mixOmics is a Bioconductor package; installation takes ~5 minutes.
# Uncomment the block below and run it, then re-run cell 7.1.

# import rpy2.robjects as ro
# ro.r("""
#     if (!requireNamespace('BiocManager', quietly = TRUE))
#         install.packages('BiocManager', repos = 'https://cloud.r-project.org')
#     if (!requireNamespace('mixOmics', quietly = TRUE))
#         BiocManager::install('mixOmics', ask = FALSE, update = FALSE)
# """)
# print('mixOmics installation complete — re-run cell 7.1 to verify.')

print('Cell 7.0: optional install cell — uncomment and run only if mixOmics is missing.')
```

```python
# ── 7.1  Setup: load mixOmics via rpy2 ───────────────────────────────────────
# R 4.5.3 and rpy2 3.6.7 are pre-installed in SysCompBio_tf.
# Install mixOmics once via cell 7.0 if DIABLO_AVAILABLE = False below.

import rpy2.robjects as ro
from rpy2.robjects import pandas2ri
from rpy2.robjects.conversion import localconverter
import rpy2.robjects.packages as rpackages

DIABLO_AVAILABLE = False
try:
    mixomics = rpackages.importr('mixOmics')
    ver = ro.r('as.character(packageVersion("mixOmics"))')[0]
    print(f'mixOmics {ver} loaded.')
    DIABLO_AVAILABLE = True
except Exception as e:
    print(f'mixOmics not available: {e}')
    print('Uncomment and run cell 7.0 to install it, then re-run this cell.')

print(f'\nDIABLO available: {DIABLO_AVAILABLE}')
```

```python
# ── 7.2  Data preparation ─────────────────────────────────────────────────────
# DIABLO requires all three modalities to share the same row order.
# Use only patients present in ALL three harmonised dataframes with metadata labels.

if not DIABLO_AVAILABLE:
    print('Skipping data prep — mixOmics not available (run cells 7.0 and 7.1 first).')
else:
    # Dedup to one measurement per patient before computing the intersection
    # (DNAm can have multiple tumor measurements per patient after normalisation)
    rna_dd  = rnaseq[~rnaseq.index.duplicated(keep='first')]
    dnam_dd = dnam[~dnam.index.duplicated(keep='first')]
    cnv_dd  = cnv[~cnv.index.duplicated(keep='first')]

    # Intersection of patients with all 3 modalities + metadata label
    shared = (rna_dd.index
              .intersection(dnam_dd.index)
              .intersection(cnv_dd.index))
    lbl_shared = pd.Series(shared.map(label_map), index=shared)
    shared = shared[~lbl_shared.isna()]
    y_shared = lbl_shared.dropna().astype(int).values

    print(f'Shared patients (all 3 modalities + metadata): {len(shared)}')
    print(f'  Normal (0): {(y_shared == 0).sum()}   Tumour (1): {(y_shared == 1).sum()}')

    # Variance filter — same N_PAIR features as pairwise experiments
    rna_f  = select_top_var(rna_dd.loc[shared],  N_PAIR)
    dnam_f = select_top_var(dnam_dd.loc[shared], N_PAIR)
    cnv_f  = select_top_var(cnv_dd.loc[shared],  N_PAIR)

    print(f'\nFeatures per block after variance filter (top {N_PAIR}):')
    print(f'  RNASeq : {rna_f.shape[1]}')
    print(f'  DNAm   : {dnam_f.shape[1]}')
    print(f'  CNV    : {cnv_f.shape[1]}')

    # Convert DataFrames to R matrices
    with localconverter(ro.default_converter + pandas2ri.converter):
        r_rna  = ro.conversion.py2rpy(rna_f)
        r_dnam = ro.conversion.py2rpy(dnam_f)
        r_cnv  = ro.conversion.py2rpy(cnv_f)

    # R factor: labels as 'Normal' / 'Tumour'
    y_labels = ['Normal' if v == 0 else 'Tumour' for v in y_shared]
    r_Y = ro.FactorVector(ro.StrVector(y_labels))

    # Push to R global environment
    ro.globalenv['r_rna']  = r_rna
    ro.globalenv['r_dnam'] = r_dnam
    ro.globalenv['r_cnv']  = r_cnv
    ro.globalenv['r_Y']    = r_Y
    ro.r('X_list <- list(RNASeq = r_rna, DNAm = r_dnam, CNV = r_cnv)')

    print('\nR objects created: X_list (3 blocks) and Y factor — ready for DIABLO.')
```

```python
# ── 7.3  Fit DIABLO and estimate performance ──────────────────────────────────
# Design matrix: off-diagonal 0.1 → moderate cross-block covariance.
# ncomp = 2 latent components; keepX = 50 features per block per component.
# perf(): 3-fold × 10-repeat CV — 3 folds because only 3 Normal samples exist
# in the shared intersection (same logic as the sklearn pipeline's auto-reduce).

diablo_perf = None

if not DIABLO_AVAILABLE:
    print('Skipping DIABLO fit — mixOmics not available.')
else:
    print('Fitting DIABLO (block.splsda, ncomp=2, keepX=50 per block) ...')
    ro.r("""
        design <- matrix(0.1,
            nrow = length(X_list), ncol = length(X_list),
            dimnames = list(names(X_list), names(X_list)))
        diag(design) <- 0

        diablo_fit <- block.splsda(
            X      = X_list,
            Y      = r_Y,
            ncomp  = 2,
            keepX  = list(RNASeq = c(50, 50), DNAm = c(50, 50), CNV = c(50, 50)),
            design = design
        )
    """)
    print('Model fitted.')
    print('Running cross-validation (3 folds × 10 repeats) — this takes a few minutes ...')

    ro.r("""
        set.seed(42)
        diablo_perf_obj <- perf(
            diablo_fit,
            validation = 'Mfold',
            folds      = 3,
            nrepeat    = 10,
            auc        = TRUE
        )
    """)
    diablo_perf = ro.globalenv['diablo_perf_obj']
    print('Cross-validation complete.')
```

```python
# ── 7.4  Results: DIABLO vs sklearn pipeline ─────────────────────────────────
# perf() object structure (mixOmics 6.x):
#   auc$comp2[, "AUC"]                                — AUC per class pair at comp 2
#   AveragedPredict.error.rate["Overall.BER", "comp2"] — balanced error rate at comp 2

if not DIABLO_AVAILABLE or diablo_perf is None:
    print('Skipping results — DIABLO not run (run cells 7.1–7.3 first).')
else:
    diablo_auc = float(ro.r(
        'mean(diablo_perf_obj$auc$comp2[, "AUC"])'
    )[0])
    diablo_ber = float(ro.r(
        'diablo_perf_obj$AveragedPredict.error.rate["Overall.BER", "comp2"]'
    )[0])
    diablo_bal_acc = round(1.0 - diablo_ber, 4)
    diablo_auc     = round(diablo_auc, 4)

    print(f'DIABLO (averaged predictions, comp 2):')
    print(f'  AUC-ROC      : {diablo_auc}')
    print(f'  Balanced Acc : {diablo_bal_acc}  (= 1 − BER, AveragedPredict)')

    diablo_metrics = {
        'AUC-ROC':      diablo_auc,
        'Balanced Acc': diablo_bal_acc,
        'F1 (macro)':   float('nan'),
        'Accuracy':     float('nan'),
    }
    diablo_row = pd.DataFrame([{
        'Dataset': 'DIABLO (all 3)', 'Model': 'block.splsda', **diablo_metrics
    }])

    # Best model per configuration from the sklearn pipeline
    best_pipeline = (
        all_summary
        .sort_values('AUC-ROC', ascending=False)
        .groupby('Dataset', sort=False)
        .first()[['Model'] + METRIC_COLS]
        .reset_index()
        .reindex(columns=['Dataset', 'Model'] + METRIC_COLS)
    )
    compare_df = pd.concat([best_pipeline, diablo_row], ignore_index=True)

    print('\n=== DIABLO vs best sklearn-pipeline model ===')
    print('Note: sklearn metrics = 5-fold CV mean; DIABLO = 3-fold × 10-repeat CV.\n')
    display(compare_df.set_index('Dataset').style
            .format('{:.3f}', na_rep='—', subset=METRIC_COLS)
            .highlight_max(subset=['AUC-ROC', 'Balanced Acc'], axis=0, color='#c6efce'))

    # Bar chart — best AUC-ROC per config including DIABLO
    cfg_order = list(MODALITY_ORDER) + ['DIABLO (all 3)']
    auc_vals  = []
    for cfg in cfg_order:
        if cfg == 'DIABLO (all 3)':
            auc_vals.append(diablo_auc)
        else:
            auc_vals.append(max(all_results[cfg][m]['AUC-ROC'] for m in models_list))

    colors = [PALETTE.get(c, '#b07aa1') for c in cfg_order]
    fig, ax = plt.subplots(figsize=(10, 4))
    bars = ax.bar(cfg_order, auc_vals, color=colors, alpha=0.85, edgecolor='white')
    ax.set_ylim(0, 1.1)
    ax.axhline(0.5, ls='--', color='grey', lw=0.9, alpha=0.5, label='Random baseline')
    ax.axvline(5.5, ls=':', color='black', lw=1, alpha=0.4)
    ax.annotate('← sklearn pipeline', xy=(0.01, 1.04), xycoords='axes fraction',
                fontsize=8, color='dimgrey')
    ax.annotate('DIABLO →', xy=(0.87, 1.04), xycoords='axes fraction',
                fontsize=8, color='dimgrey')
    for bar, val in zip(bars, auc_vals):
        if not np.isnan(val):
            ax.text(bar.get_x() + bar.get_width() / 2, val + 0.01, f'{val:.3f}',
                    ha='center', va='bottom', fontsize=8)
    ax.set_ylabel('AUC-ROC (best model per config)')
    ax.set_xticklabels(cfg_order, rotation=20, ha='right', fontsize=9)
    ax.set_title('Best AUC-ROC per configuration — sklearn pipeline vs DIABLO (all 3 layers)',
                 fontsize=12)
    ax.legend(fontsize=8)
    plt.tight_layout()
    plt.show()
```

   - Early fusion ignores modality-specific structure; DIABLO addresses this via the
     design matrix, but sparse PLS-DA still imposes linearity assumptions.
   - Feature selection by variance is unsupervised and may not pick the most
     discriminative features.
   - Pairwise combinations use fewer samples (intersection only), which can affect
     model stability.
   - DIABLO's F1 and raw Accuracy are not directly extracted from `perf()`;
     comparison to sklearn models on those metrics is therefore incomplete.

## 9. Biological Interpretation — SHAP Analysis (XGBoost)

**Why SHAP?**  
SHAP (SHapley Additive exPlanations) assigns each feature a contribution score per prediction.

**PCA backprojection for RNA and DNAm:**

Because XGBoost operates on PCA components, SHAP values directly measure *component importance*.
To recover *gene-level importance*, we project back through the PCA loading matrix:

$$\text{importance}_j = \sum_{k=1}^{K} \overline{|\text{SHAP}|}_k \cdot |W_{kj}|$$

where $\overline{|\text{SHAP}|}_k$ is the mean absolute SHAP value for PC $k$ and $W_{kj}$ is the loading
of gene $j$ on PC $k$. This is a single matrix multiply — computationally trivial.

**Summary of interpretability per modality:**
| Modality | SHAP features | Backprojection | Result |
|---------|--------------|---------------|--------|
| RNASeq | PCA components | ✓ direct | Named genes |
| DNAm | PCA components | ✓ direct | CpG probe IDs (need manifest → genes) |
| CNV | Raw genes (no PCA) | not needed | Named genes |

**Sections:**
- **8.1** — RNASeq: SHAP on components + backprojection to genes
- **8.2** — DNAm: SHAP on components + backprojection to CpG probes
- **8.3** — CNV: SHAP directly on gene-level features
- **8.4** — Cross-fold Jaccard stability (CNV)
- **Section 9** — Pairwise fusion SHAP with backprojection
- **Section 10** — KEGG/GO enrichment (RNASeq and CNV genes)

> **Requires:** `pip install shap`

```python
import shap

# Helper: project component-level SHAP back to original feature space
def shap_backproject(mean_abs_shap_per_component, pca):
    """
    mean_abs_shap_per_component : array of shape (n_components,)
    pca                         : fitted sklearn PCA object
    Returns                     : array of shape (n_features,) — one importance score per input feature
    """
    return np.abs(pca.components_).T @ mean_abs_shap_per_component


# ── 8.1 RNASeq ────────────────────────────────────────────────────────────────
X_rna_full, y_rna_full = prepare_single(rnaseq, label_map, N_SINGLE)
feature_names_rna = select_top_var(rnaseq, N_SINGLE).columns.tolist()

xgb_rna = clone(make_single_models(use_pca=True)['XGBoost'])
xgb_rna.fit(X_rna_full, y_rna_full)

X_rna_pca = xgb_rna[:-1].transform(X_rna_full)
feature_names_rna_pca = [f'PC{i}' for i in range(X_rna_pca.shape[1])]

explainer_rna = shap.TreeExplainer(xgb_rna.named_steps['clf'])
shap_values_rna = explainer_rna.shap_values(X_rna_pca)
mean_shap_rna = np.abs(shap_values_rna).mean(axis=0)

# --- Component-level plot ---
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

plt.sca(axes[0])
shap.summary_plot(shap_values_rna, X_rna_pca,
                  feature_names=feature_names_rna_pca, max_display=15, show=False)
axes[0].set_title('RNASeq — SHAP on PCA components')

# --- Backprojection to gene space ---
gene_imp_rna = shap_backproject(mean_shap_rna, xgb_rna.named_steps['pca'])
shap_gene_rna = (pd.DataFrame({'gene': feature_names_rna, 'importance': gene_imp_rna})
                   .sort_values('importance', ascending=False)
                   .reset_index(drop=True))

top_genes_plot = shap_gene_rna.head(20)
axes[1].barh(top_genes_plot['gene'][::-1], top_genes_plot['importance'][::-1],
             color=PALETTE['RNASeq'], alpha=0.85)
axes[1].set_xlabel('SHAP-weighted loading importance')
axes[1].set_title('RNASeq — top 20 genes (SHAP backprojected through PCA)')
axes[1].tick_params(axis='y', labelsize=8)

plt.suptitle('RNASeq XGBoost SHAP — component view and gene-level backprojection', fontsize=12)
plt.tight_layout()
plt.show()

print('Top 20 RNASeq genes by SHAP-PCA backprojection:')
display(shap_gene_rna.head(20))
```

```python
# 9.4 Cross-fold SHAP stability for CNV (gene-level, interpretable)
# RNA/DNAm use PCA components — Jaccard on component indices is less meaningful biologically.
top_n = 30

def compute_fold_jaccard(X_full, y_full, use_pca, top_n=30):
    fold_top = []
    skf_shap = StratifiedKFold(n_splits=CV_FOLDS, shuffle=True, random_state=SEED)
    for train_idx, test_idx in skf_shap.split(X_full, y_full):
        pipe = clone(make_single_models(use_pca=use_pca)['XGBoost'])
        pipe.fit(X_full[train_idx], y_full[train_idx])
        X_test_transformed = pipe[:-1].transform(X_full[test_idx])
        exp = shap.TreeExplainer(pipe.named_steps['clf'])
        sv  = exp.shap_values(X_test_transformed)
        top_idx = set(np.argsort(np.abs(sv).mean(axis=0))[-top_n:])
        fold_top.append(top_idx)
    return fold_top

def plot_jaccard(fold_top, title, ax):
    n   = len(fold_top)
    jac = np.zeros((n, n))
    for i in range(n):
        for j in range(n):
            inter = len(fold_top[i] & fold_top[j])
            union = len(fold_top[i] | fold_top[j])
            jac[i, j] = inter / union if union > 0 else 1.0
    sns.heatmap(jac, annot=True, fmt='.2f', cmap='Blues', vmin=0, vmax=1,
                xticklabels=[f'Fold {i+1}' for i in range(n)],
                yticklabels=[f'Fold {i+1}' for i in range(n)], ax=ax)
    ax.set_title(title)
    return jac[np.triu_indices(n, k=1)].mean()

fig, ax = plt.subplots(1, 1, figsize=(6, 5))

print('Computing cross-fold Jaccard for CNV (gene-level SHAP) ...')
fold_top_cnv = compute_fold_jaccard(X_cnv_full, y_cnv_full, use_pca=False, top_n=top_n)
jac_cnv = plot_jaccard(fold_top_cnv,
    f'Jaccard similarity — top-{top_n} SHAP genes\n(CNV, XGBoost, no PCA)', ax)

plt.suptitle(f'Cross-fold SHAP feature stability — top-{top_n} CNV genes, 5-fold CV', fontsize=13)
plt.tight_layout()
plt.show()

print(f'Mean cross-fold Jaccard — CNV: {jac_cnv:.3f}')
print('(> 0.5 = good stability; > 0.7 = strong stability)')
print('\nNote: Jaccard stability for RNASeq/DNAm is not shown — SHAP values')
print('index PCA components, so fold consistency of component indices is less')
```

Each pairwise XGBoost uses a `ColumnTransformer` that applies the same PCA policy as in CV:
- **RNASeq+DNAm**: both blocks PCA-compressed → SHAP maps to PCA components.
- **RNASeq+CNV**: RNA block PCA-compressed, CNV block raw → CNV side maps to gene names.

For combinations with a raw CNV block, per-modality SHAP contributions are computed to
quantify which data layer dominates the fusion model.

## 11. KEGG / GO Pathway Enrichment

Top SHAP genes from gene-annotated modalities (RNASeq, CNV) are tested for pathway enrichment
using the Enrichr API (requires internet access).

**Note on DNAm:** Features are CpG probe IDs (e.g. `cg21870274`), not gene names. Pathway enrichment
requires mapping probes to genes via the Illumina 450K/EPIC manifest — not performed here.

> **Requires:** `pip install gseapy`

**Interpretation guide:**
- Low adj. p-value in oncogenic pathways (*p53*, *EGFR*, *KRAS*) → model captures known LUAD biology.
- Cell-cycle / DNA repair enrichment from CNV features → copy-number instability driving tumour identity.

## 11.1 DNAm: CpG → Gene Mapping via Illumina 450K Manifest + `mygene`

**Why two steps?**

`mygene` is a gene query service — it cannot directly look up CpG probe IDs. The probe-to-gene
assignment comes from the **Illumina HumanMethylation450 manifest**, which annotates every probe
with the UCSC RefGene names of overlapping or nearby genes (`UCSC_RefGene_Name` column,
semicolon-separated). One probe can overlap multiple gene bodies/isoforms, so the column often
contains several genes.

**Pipeline:**

1. Download the manifest (first run only, ~30 MB compressed).
2. Map top-SHAP CpG probes → gene lists; deduplicate repeated transcripts.
3. Pass all raw gene symbols through **`mygene.querymany`** → resolves aliases, retired symbols,
   and capitalisation variants to current official HGNC symbols.
4. Aggregate SHAP importance per gene (sum over all CpGs that map to it) → ranked gene list.
5. Run Enrichr on the ranked gene list exactly as for RNASeq / CNV.

> **Requires:** `pip install mygene`  
> The manifest is cached locally as `hm450_manifest_slim.csv` after the first download.

```python
# !pip install mygene   # uncomment if not installed

import os
import gzip
import pandas as pd
import numpy as np
import mygene

MANIFEST_FILE  = "hm450_manifest.csv.gz"   # manually downloaded file; despite .gz name, yours is plain CSV
MANIFEST_CACHE = "hm450_manifest_slim.csv"

def is_gzip_file(path):
    with open(path, "rb") as f:
        return f.read(2) == b"\x1f\x8b"

# -- Step 1: load and slim the manifest (keep only probe ID + gene columns) --
if not os.path.exists(MANIFEST_CACHE):
    if not os.path.exists(MANIFEST_FILE):
        raise FileNotFoundError(
            f"Could not find {MANIFEST_FILE}. Put the downloaded HM450 manifest in the group project folder."
        )

    compression = "gzip" if is_gzip_file(MANIFEST_FILE) else None
    print(f"Parsing manifest from {MANIFEST_FILE} ...")

    manifest_full = pd.read_csv(
        MANIFEST_FILE,
        compression=compression,
        skiprows=7,          # first 7 rows are Illumina file-header metadata
        low_memory=False,
        usecols=["Name", "UCSC_RefGene_Name"],
    )

    manifest_full = manifest_full[manifest_full["Name"].str.startswith("cg", na=False)]
    manifest_full.columns = ["cpg", "genes_raw"]
    manifest_full.to_csv(MANIFEST_CACHE, index=False)

    print(f"Manifest cached to {MANIFEST_CACHE} ({len(manifest_full):,} probes)")
else:
    manifest_full = pd.read_csv(MANIFEST_CACHE)
    print(f"Loaded cached manifest ({len(manifest_full):,} probes)")

# -- Step 2: map CpG probes to gene lists --
manifest_idx = manifest_full.set_index("cpg")["genes_raw"]

TOP_N_CPG = 500

cpg_gene_map = {}
for cpg in shap_cpg_dnam.head(TOP_N_CPG)["cpg"]:
    raw = manifest_idx.get(cpg, "")

    if pd.isna(raw) or raw == "":
        cpg_gene_map[cpg] = []
    else:
        seen, unique = set(), []
        for g in str(raw).split(";"):
            g = g.strip()
            if g and g not in seen:
                seen.add(g)
                unique.append(g)
        cpg_gene_map[cpg] = unique

n_mapped = sum(1 for v in cpg_gene_map.values() if v)
print(f"{n_mapped}/{TOP_N_CPG} CpG probes map to at least one gene")

genes_per_cpg = [len(v) for v in cpg_gene_map.values() if v]
if genes_per_cpg:
    print(f"Genes per probe: median={np.median(genes_per_cpg):.0f}, max={max(genes_per_cpg)}")
else:
    print("No CpG probes mapped to genes.")

# -- Step 3: standardise gene symbols with mygene --
all_raw_genes = sorted({g for genes in cpg_gene_map.values() for g in genes})
print(f"\nQuerying mygene for {len(all_raw_genes)} unique gene symbols ...")

symbol_std = {}

if all_raw_genes:
    mg = mygene.MyGeneInfo()

    results = mg.querymany(
        all_raw_genes,
        scopes="symbol,alias",
        fields="symbol",
        species="human",
        returnall=False,
        verbose=False,
    )

    for r in results:
        if not r.get("notfound") and "symbol" in r:
            symbol_std[r["query"]] = r["symbol"]

n_resolved = sum(1 for g in all_raw_genes if g in symbol_std)
print(f"mygene resolved {n_resolved}/{len(all_raw_genes)} symbols to current HGNC names")

# Apply standardisation; deduplicate within each probe's gene list
cpg_gene_map_std = {}

for cpg, genes in cpg_gene_map.items():
    seen, unique = set(), []
    for g in genes:
        std = symbol_std.get(g, g)
        if std not in seen:
            seen.add(std)
            unique.append(std)
    cpg_gene_map_std[cpg] = unique

print("\nExample mappings (first 5 probes with genes):")
shown = 0
for cpg, genes in cpg_gene_map_std.items():
    if genes:
        print(f"  {cpg}: {genes}")
        shown += 1
        if shown >= 5:
            break
```

```python
# ── Step 4: aggregate SHAP importance per gene ──
# A gene gets credit from every CpG that maps to it, weighted by that CpG's SHAP importance.

cpg_shap = shap_cpg_dnam.head(TOP_N_CPG).set_index('cpg')['importance']

gene_shap: dict[str, float] = {}
for cpg, genes in cpg_gene_map_std.items():
    imp = cpg_shap.get(cpg, 0.0)
    for g in genes:
        gene_shap[g] = gene_shap.get(g, 0.0) + imp

shap_gene_dnam = (
    pd.DataFrame(gene_shap.items(), columns=['gene', 'shap_weight'])
    .sort_values('shap_weight', ascending=False)
    .reset_index(drop=True)
)

print(f'Unique genes after CpG→gene mapping: {len(shap_gene_dnam)}')

# --- Bar chart: top 20 genes ---
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

top20_dnam_genes = shap_gene_dnam.head(20)
axes[0].barh(top20_dnam_genes['gene'][::-1], top20_dnam_genes['shap_weight'][::-1],
             color=PALETTE['DNAm'], alpha=0.85)
axes[0].set_xlabel('Cumulative SHAP importance (summed over mapped CpGs)')
axes[0].set_title('DNAm — top 20 genes (CpG→gene, mygene-standardised)')
axes[0].tick_params(axis='y', labelsize=8)

# --- Histogram: genes per CpG ---
axes[1].hist(genes_per_cpg, bins=range(0, max(genes_per_cpg) + 2), color=PALETTE['DNAm'],
             alpha=0.8, edgecolor='white')
axes[1].set_xlabel('Number of genes per CpG probe')
axes[1].set_ylabel('Count')
axes[1].set_title('Distribution of genes per CpG probe\n(all mapped probes in top-SHAP set)')

plt.suptitle('DNAm CpG → gene mapping (Illumina 450K manifest + mygene)', fontsize=12)
plt.tight_layout()
plt.show()

print('Top 20 genes by SHAP-weighted CpG importance:')
display(shap_gene_dnam.head(20))

# ── Step 5: enrichment analysis ──
# Use the ranked gene list (top 100 by SHAP weight) for Enrichr.
# All genes from multi-mapped CpGs are included — no single-gene-per-probe restriction.
dnam_gene_list = shap_gene_dnam['gene'].head(100).tolist()
sig_dnam = run_enrichr(dnam_gene_list, 'DNAm — CpG→gene (manifest + mygene, SHAP-weighted)')
if len(sig_dnam):
    display(sig_dnam[['Gene_set', 'Term', 'Overlap', 'P-value', 'Adjusted P-value', 'Genes']].head(20))
else:
    print('No significant pathways (adj. p < 0.05) — try relaxing the threshold or using more genes.')
```

```python
# ── 12.2  Cross-fold / cross-seed feature stability (RF, no PCA) ───────────────
print(f'Running stability analysis  ({len(STAB_SEEDS)} seeds × {STAB_FOLDS} folds, '
      f'100 trees per fold) ...')
stab_results = {}

for name in MODS:
    d  = interp_data[name]
    X, y, fn = d['X'], d['y'], d['feat_names']
    n_splits  = min(STAB_FOLDS, np.bincount(y).min())

    fold_top_sets  = []
    fold_imps_list = []

    for seed in STAB_SEEDS:
        skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=seed)
        for train_idx, _ in skf.split(X, y):
            rf = RandomForestClassifier(n_estimators=100, max_features='sqrt',
                                        class_weight='balanced',
                                        random_state=seed, n_jobs=-1)
            rf.fit(X[train_idx], y[train_idx])
            imp = rf.feature_importances_
            fold_imps_list.append(imp)
            fold_top_sets.append(set(np.argsort(imp)[-TOP_K:]))

    n_runs  = len(fold_top_sets)
    all_idx = set().union(*fold_top_sets)
    counts  = {i: sum(1 for s in fold_top_sets if i in s) for i in all_idx}
    stable  = sorted([i for i, c in counts.items() if c >= 0.8 * n_runs],
                     key=lambda i: -counts[i])

    stab_results[name] = {
        'fold_imps': np.array(fold_imps_list),
        'counts':    counts,
        'stable':    stable,
        'n_runs':    n_runs,
        'n_splits':  n_splits,
    }
    top5 = list(fn[stable[:5]]) if stable else []
    print(f'  {name}: {len(stable):3d} stable features  (top-5: {top5})')

# ── Heatmap: importance per fold-seed run for stable features ─────────────────
fig, axes = plt.subplots(1, 3, figsize=(20, 9))

for ax, name in zip(axes, MODS):
    r   = stab_results[name]
    fn  = interp_data[name]['feat_names']
    idx = r['stable'][:30]

    if not idx:
        ax.text(0.5, 0.5, 'No stable features\n(≥ 80% threshold)',
                ha='center', va='center', transform=ax.transAxes, fontsize=11)
        ax.set_title(name)
        continue

    mat  = r['fold_imps'][:, idx]
    # Column-normalise: each feature scaled to its own max across runs
    norm = mat / (mat.max(axis=0, keepdims=True) + 1e-10)

    n_splits = r['n_splits']
    rl = [f'S{s}F{f+1}' for s in range(len(STAB_SEEDS)) for f in range(n_splits)]

    sns.heatmap(norm.T, ax=ax, cmap='YlOrRd',
                yticklabels=fn[idx], xticklabels=rl,
                cbar_kws={'label': 'Rel. importance'})
    ax.set_title(f'{name} — {len(idx)} stable features\n'
                 f'(≥80 % of {r["n_runs"]} fold×seed runs)', fontsize=9)
    ax.tick_params(axis='y', labelsize=6)
    ax.tick_params(axis='x', labelsize=7, rotation=45)

plt.suptitle(f'RF feature stability — {len(STAB_SEEDS)} seeds × {STAB_FOLDS} folds, '
             f'100 trees, no PCA\n(column-normalised importance; rows = features, '
             f'cols = fold×seed runs)',
             fontsize=12)
plt.tight_layout()
plt.show()
```

```python
# ── 12.3  Cross-model feature-rank consistency ─────────────────────────────────
fig, axes = plt.subplots(1, 3, figsize=(14, 4))

for ax, name in zip(axes, MODS):
    fn = interp_data[name]['feat_names']
    m  = interp_models[name]
    ranks = pd.DataFrame({
        lbl: pd.Series(m[key], index=fn).rank(ascending=False)
        for key, lbl in zip(IMP_KEYS, IMP_LABELS)
    })
    corr = ranks[(ranks <= 200).any(axis=1)].corr(method='spearman')
    sns.heatmap(corr, annot=True, fmt='.3f', cmap='coolwarm', vmin=0, vmax=1,
                ax=ax, square=True, linewidths=0.5, cbar_kws={'label': 'Spearman \u03c1'})
    ax.set_title(f'{name}', fontsize=10, color=PALETTE[name], fontweight='bold')

model_desc = '  \u00b7  '.join(IMP_LABELS)
plt.suptitle(f'Cross-model feature-rank consistency — Spearman \u03c1\n'
             f'({model_desc}, top-200 by any model, no PCA)',
             fontsize=12)
plt.tight_layout()
plt.show()

# ── Consensus: features in top-K for ALL model families ──────────────────────
print(f'\nConsensus features — top-{TOP_K} in ALL {len(IMP_KEYS)} model families:')
for name in MODS:
    fn = interp_data[name]['feat_names']
    m  = interp_models[name]
    sets = [set(fn[np.argsort(m[k])[-TOP_K:]]) for k in IMP_KEYS]
    cons = sets[0].intersection(*sets[1:])
    if cons:
        print(f'  {name}: {sorted(cons)}')
    else:
        sets50 = [set(fn[np.argsort(m[k])[-50:]]) for k in IMP_KEYS]
        cons50 = sets50[0].intersection(*sets50[1:])
        print(f'  {name}: (none at top-{TOP_K}; top-50 consensus: {sorted(cons50)})')
```

```python
# ── 12.4a  Pairwise attribution models (no PCA) ────────────────────────────────
# Concatenate top-N_PAIR_INTERP features from each block (with modality prefix)
# and fit all active attribution models without PCA.

N_PAIR_INTERP = 1000

PAIR_SOURCES = [
    ('RNASeq+DNAm', rnaseq, dnam, 'RNA',  'DNAm', 'RNASeq', 'DNAm'),
    ('RNASeq+CNV',  rnaseq, cnv,  'RNA',  'CNV',  'RNASeq', 'CNV'),
    ('DNAm+CNV',    dnam,   cnv,  'DNAm', 'CNV',  'DNAm',   'CNV'),
]
PAIR_NAMES = [p[0] for p in PAIR_SOURCES]

pair_interp_data   = {}   # pair_name → {feat_names, X, y, n1, n2, p1, p2}
pair_interp_models = {}   # pair_name → {key: importance_array, ...}

for pair_name, df1, df2, p1, p2, n1, n2 in PAIR_SOURCES:
    print(f'Fitting pairwise attribution models ({pair_name}) ...', end=' ', flush=True)

    sel1 = select_top_var(df1, N_PAIR_INTERP).add_prefix(p1 + '_')
    sel2 = select_top_var(df2, N_PAIR_INTERP).add_prefix(p2 + '_')
    sel1 = sel1[~sel1.index.duplicated(keep='first')]
    sel2 = sel2[~sel2.index.duplicated(keep='first')]
    shared = sel1.index.intersection(sel2.index)

    X_df  = pd.concat([sel1.loc[shared], sel2.loc[shared]], axis=1)
    lbl   = X_df.index.map(label_map)
    valid = ~lbl.isna()
    X_df  = X_df.loc[valid]
    y     = lbl[valid].astype(int).values
    X     = StandardScaler().fit_transform(X_df.values.astype(float))

    imp_dict = {}
    for mname, key, slbl, llbl, factory in _active:
        clf = factory()
        clf.fit(X, y)
        imp_dict[key] = (clf.feature_importances_
                         if hasattr(clf, 'feature_importances_')
                         else np.abs(clf.coef_[0]))

    pair_interp_data[pair_name] = {
        'feat_names': np.array(X_df.columns), 'X': X, 'y': y,
        'n1': n1, 'n2': n2, 'p1': p1, 'p2': p2,
    }
    pair_interp_models[pair_name] = imp_dict
    print(f'n={len(y)}  Normal={(y==0).sum()}  Tumour={(y==1).sum()}')

print(f'\nAll pairwise attribution models fitted ({len(IMP_LABELS)} families: {IMP_LABELS}).')
```

```python
# ── 12.4  Pathway enrichment (Enrichr) ────────────────────────────────────────
# RNASeq and CNV: gene symbols → Enrichr.
# DNAm skipped: CpG IDs need annotation (Illumina 450k manifest) to map to genes.

enrich_results = {}

if not GSEAPY_AVAILABLE:
    print('gseapy not available — install with: pip install gseapy')
else:
    for name in ['RNASeq', 'CNV']:
        fn = interp_data[name]['feat_names']
        m  = interp_models[name]
        r  = stab_results[name]

        # Use stable features if ≥10 found, otherwise fall back to RF top-N
        if len(r['stable']) >= 10:
            gene_list = list(fn[r['stable']])[:N_ENRICH]
            src_tag   = f'{len(gene_list)} stable features'
        else:
            gene_list = list(fn[np.argsort(m[IMP_KEYS[0]])[-N_ENRICH:][::-1]])
            src_tag   = f'RF top-{N_ENRICH}'

        print(f'{name}: querying Enrichr ({src_tag}) ...')
        enr_dfs = []
        for db in ENRICH_DBS:
            try:
                enr = gp.enrichr(gene_list=gene_list, gene_sets=db,
                                  outdir=None, verbose=False)
                enr_dfs.append(enr.results)
            except Exception as e:
                print(f'  {db}: {e}')
        if enr_dfs:
            combined = pd.concat(enr_dfs, ignore_index=True)
            enrich_results[name] = combined
            for db in ENRICH_DBS:
                sig = combined[(combined['Gene_set'] == db) &
                               (combined['Adjusted P-value'] < 0.05)]
                print(f'  {db}: {len(sig)} significant terms (adj. p < 0.05)')

# ── Plot ──────────────────────────────────────────────────────────────────────
avail = [n for n in ['RNASeq', 'CNV'] if n in enrich_results]
if avail:
    fig, axes = plt.subplots(len(avail), len(ENRICH_DBS),
                             figsize=(21, 5 * len(avail)))
    if len(avail) == 1:
        axes = axes[np.newaxis, :]

    for ri, name in enumerate(avail):
        res = enrich_results[name]
        for ci, db in enumerate(ENRICH_DBS):
            ax  = axes[ri][ci]
            sub = (res[res['Gene_set'] == db]
                   .sort_values('Adjusted P-value')
                   .head(10))
            sig = sub[sub['Adjusted P-value'] < 0.05]

            if sig.empty:
                ax.text(0.5, 0.5, 'No significant terms\n(adj. p ≥ 0.05)',
                        ha='center', va='center', transform=ax.transAxes, fontsize=9)
            else:
                colors = plt.cm.YlOrRd_r(np.linspace(0.15, 0.85, len(sig)))
                ax.barh(range(len(sig)), -np.log10(sig['Adjusted P-value'].values),
                        color=colors)
                ax.set_yticks(range(len(sig)))
                ax.set_yticklabels([t[:55] for t in sig['Term']], fontsize=7)
                ax.axvline(-np.log10(0.05), ls='--', color='grey', lw=0.8)
                ax.set_xlabel('-log₁₀(adj. p)')
            ax.set_title(f'{name} × {db}', fontsize=8)
            if ri == 0 and ci == 0:
                ax.set_ylabel(name, fontsize=9)

    plt.suptitle('Pathway enrichment — top stable features submitted to Enrichr\n'
                 'DNAm excluded (CpG IDs require gene annotation to map)',
                 fontsize=12)
    plt.tight_layout()
    plt.show()
elif GSEAPY_AVAILABLE:
    print('No enrichment results (all Enrichr queries failed).')

# ── Pairwise enrichment ───────────────────────────────────────────────────────
# Genes in top-K for ≥ 2 model families in any pairwise combination,
# submitted to Enrichr.  Answers: does multi-omics fusion surface new pathways?

if GSEAPY_AVAILABLE and pair_interp_data:
    pair_gene_candidates = {}
    for pair_name in PAIR_NAMES:
        fn  = pair_interp_data[pair_name]['feat_names']
        m   = pair_interp_models[pair_name]
        top_per_model = []
        for imp_key in IMP_KEYS:
            top_feats = fn[np.argsort(m[imp_key])[-TOP_K:]]
            genes = {f.split('_', 1)[-1] for f in top_feats
                     if not f.split('_', 1)[-1].startswith('cg')}
            top_per_model.append(genes)
        # genes appearing in top-K for at least 2 models
        pairwise_cons = set().union(*(
            top_per_model[i] & top_per_model[j]
            for i in range(len(IMP_KEYS))
            for j in range(i + 1, len(IMP_KEYS))
        ))
        pair_gene_candidates[pair_name] = pairwise_cons

    pair_gene_list = sorted(set.union(*pair_gene_candidates.values()))[:N_ENRICH]
    print(f'\nPairwise consensus gene list: {len(pair_gene_list)} genes '
          f'(top-{TOP_K} in ≥2 models, any pair, CpGs excluded)')

    pair_enr_dfs = []
    for db in ENRICH_DBS:
        try:
            enr = gp.enrichr(gene_list=pair_gene_list, gene_sets=db,
                              outdir=None, verbose=False)
            pair_enr_dfs.append(enr.results)
        except Exception as e:
            print(f'  {db}: {e}')

    if pair_enr_dfs:
        pair_combined = pd.concat(pair_enr_dfs, ignore_index=True)
        for db in ENRICH_DBS:
            sig = pair_combined[(pair_combined['Gene_set'] == db) &
                                (pair_combined['Adjusted P-value'] < 0.05)]
            print(f'  {db}: {len(sig)} significant terms')

        fig, axes = plt.subplots(1, len(ENRICH_DBS), figsize=(21, 5))
        for ci, db in enumerate(ENRICH_DBS):
            ax  = axes[ci]
            sub = (pair_combined[pair_combined['Gene_set'] == db]
                   .sort_values('Adjusted P-value').head(10))
            sig = sub[sub['Adjusted P-value'] < 0.05]
            if sig.empty:
                ax.text(0.5, 0.5, 'No significant terms\n(adj. p ≥ 0.05)',
                        ha='center', va='center', transform=ax.transAxes, fontsize=9)
            else:
                colors = plt.cm.YlOrRd_r(np.linspace(0.15, 0.85, len(sig)))
                ax.barh(range(len(sig)), -np.log10(sig['Adjusted P-value'].values),
                        color=colors)
                ax.set_yticks(range(len(sig)))
                ax.set_yticklabels([t[:55] for t in sig['Term']], fontsize=7)
                ax.axvline(-np.log10(0.05), ls='--', color='grey', lw=0.8)
                ax.set_xlabel('-log₁₀(adj. p)')
            ax.set_title(f'Pairwise × {db}', fontsize=8)
        plt.suptitle(f'Pathway enrichment — pairwise consensus genes (n={len(pair_gene_list)})\n'
                     f'Genes in top-{TOP_K} for ≥2 of {len(IMP_LABELS)} models ({" / ".join(IMP_LABELS)}) in any pair',
                     fontsize=12)
        plt.tight_layout()
        plt.show()
elif GSEAPY_AVAILABLE:
    print('Run section 9.4a first.')
```

```python
# ── 12.5  DIABLO sparse loadings ───────────────────────────────────────────────
diablo_loadings = {}

if not DIABLO_AVAILABLE or diablo_perf is None:
    print('Skipping DIABLO loadings — run section 7 first.')
else:
    print('Extracting sparse DIABLO loadings ...')
    for block in ['RNASeq', 'DNAm', 'CNV']:
        try:
            feat_r  = list(ro.r(f'rownames(diablo_fit$loadings${block})'))
            load_c1 = np.array(ro.r(f'diablo_fit$loadings${block}[, 1]'))
            load_c2 = np.array(ro.r(f'diablo_fit$loadings${block}[, 2]'))
            diablo_loadings[block] = {
                'comp1': sorted([(f, v) for f, v in zip(feat_r, load_c1) if v != 0],
                                key=lambda x: -abs(x[1])),
                'comp2': sorted([(f, v) for f, v in zip(feat_r, load_c2) if v != 0],
                                key=lambda x: -abs(x[1])),
            }
            c1n = len(diablo_loadings[block]['comp1'])
            c2n = len(diablo_loadings[block]['comp2'])
            print(f'  {block}: comp1={c1n}  comp2={c2n}')
        except Exception as e:
            print(f'  {block}: {e}')

if diablo_loadings:
    fig, axes = plt.subplots(2, 3, figsize=(19, 12))
    for col, block in enumerate(['RNASeq', 'DNAm', 'CNV']):
        if block not in diablo_loadings:
            continue
        for row, comp in enumerate(['comp1', 'comp2']):
            ax    = axes[row][col]
            items = diablo_loadings[block][comp][:TOP_K]
            if not items:
                ax.text(0.5, 0.5, '—', ha='center', va='center', transform=ax.transAxes)
                continue
            nms = [x[0] for x in items]
            vs  = [x[1] for x in items]
            clr = [PALETTE[block] if v > 0 else '#d62728' for v in vs]
            ax.barh(range(len(nms)), vs, color=clr, alpha=0.85)
            ax.set_yticks(range(len(nms)))
            ax.set_yticklabels(nms, fontsize=6.5)
            ax.axvline(0, color='black', lw=0.5)
            ax.set_xlabel('Loading')
            if row == 0:
                ax.set_title(block, fontsize=11, color=PALETTE[block], fontweight='bold')
            if col == 0:
                ax.set_ylabel(f'Component {row + 1}', fontsize=9)

    plt.suptitle(f'DIABLO sparse loadings — top {TOP_K} per block per component\n'
                 '(keepX = 50; positive loading → Tumour direction, negative → Normal)',
                 fontsize=12)
    plt.tight_layout()
    plt.show()
```

```python
# ── 12.6  Cross-layer / cross-model consensus ──────────────────────────────────
# Aggregate feature importance ranks from single-omics, pairwise, and DIABLO
# sources.  DNAm CpG probe IDs are excluded so only gene-name space is compared.

all_scores = {}   # {gene: {source_key: rank}}

# Single-omics (RNASeq + CNV; DNAm CpGs excluded)
for name in ['RNASeq', 'CNV']:
    fn = interp_data[name]['feat_names']
    m  = interp_models[name]
    for key, lbl in zip(IMP_KEYS, IMP_LABELS):
        ranks = pd.Series(m[key], index=fn).rank(ascending=False)
        for gene, rk in ranks.items():
            all_scores.setdefault(gene, {})[f'{name}_{lbl}'] = rk

# Pairwise (strip modality prefix; skip CpG probe IDs)
for pair_name in PAIR_NAMES:
    fn = pair_interp_data[pair_name]['feat_names']
    m  = pair_interp_models[pair_name]
    for key, lbl in zip(IMP_KEYS, IMP_LABELS):
        ranks = pd.Series(m[key], index=fn).rank(ascending=False)
        for feat, rk in ranks.items():
            gene = feat.split('_', 1)[-1]   # strip RNA_ / CNV_ / DNAm_ prefix
            if gene.startswith('cg'):        # skip methylation probes
                continue
            all_scores.setdefault(gene, {})[f'{pair_name}_{lbl}'] = rk

# DIABLO loadings
if diablo_loadings:
    for block in ['RNASeq', 'CNV']:
        if block not in diablo_loadings:
            continue
        for comp in ['comp1', 'comp2']:
            for rk_i, (gene, _) in enumerate(diablo_loadings[block][comp], start=1):
                all_scores.setdefault(gene, {})[f'{block}_DIABLO_{comp}'] = rk_i

single_cols = [f'{n}_{lbl}' for n in ['RNASeq', 'CNV'] for lbl in IMP_LABELS]
pair_cols   = [f'{pn}_{lbl}' for pn in PAIR_NAMES for lbl in IMP_LABELS]
diablo_cols = ([f'{b}_DIABLO_{c}' for b in ['RNASeq', 'CNV'] for c in ['comp1', 'comp2']]
               if diablo_loadings else [])
src_cols = single_cols + pair_cols + diablo_cols

summary = (pd.DataFrame(all_scores).T
             .reindex(columns=src_cols)
             .assign(n_top=lambda df: (df[src_cols] <= TOP_K).sum(axis=1))
             .sort_values('n_top', ascending=False))

consensus = summary[summary['n_top'] >= 2].head(40)
print(f'Genes in top-{TOP_K} across ≥ 2 model × layer sources: {len(consensus)}')
display(consensus[['n_top'] + src_cols]
        .style
        .background_gradient(cmap='YlOrRd', subset=['n_top'])
        .format('{:.0f}', na_rep='—', subset=src_cols)
        .format('{:.0f}', subset=['n_top']))

# ── Presence heatmap with group separators ────────────────────────────────────
presence = (consensus[src_cols] <= TOP_K).astype(float)
presence[consensus[src_cols].isna()] = np.nan

sep1 = len(single_cols)
sep2 = len(single_cols) + len(pair_cols)

def _fmt_col(c):
    if 'DIABLO' in c:
        parts = c.split('_')
        return f'{parts[0]}\nDIABLO {parts[2]}'
    if '+' in c:
        pn, model = c.rsplit('_', 1)
        return pn.replace('RNASeq', 'RNA') + '\n' + model
    return c.replace('_', '\n')

fig, ax = plt.subplots(figsize=(max(14, len(src_cols) * 0.6), max(5, len(consensus) * 0.38)))
sns.heatmap(presence, ax=ax, cmap='YlOrRd', vmin=0, vmax=1,
            linewidths=0.3, cbar=False,
            xticklabels=[_fmt_col(c) for c in src_cols],
            yticklabels=consensus.index)

for sep in [sep1, sep2]:
    if 0 < sep < len(src_cols):
        ax.axvline(sep, color='black', linewidth=1.8)

group_mids   = [sep1 / 2, sep1 + (sep2 - sep1) / 2, sep2 + (len(src_cols) - sep2) / 2]
group_labels = ['Single-omics', 'Pairwise', 'DIABLO']
for mid, lbl in zip(group_mids, group_labels):
    if mid < len(src_cols):
        ax.text(mid, -0.9, lbl, ha='center', va='bottom',
                fontsize=9, fontweight='bold', transform=ax.transData)

ax.set_title(f'Cross-layer / cross-model feature consensus\n'
             f'Coloured = in top-{TOP_K}  ·  grey = outside top-{TOP_K}', fontsize=11)
ax.tick_params(axis='x', labelsize=7)
ax.tick_params(axis='y', labelsize=7)
plt.tight_layout()
plt.show()
```

## 13. Network Module Analysis of Important Features

Do the top important features cluster into functionally coherent modules, and do those modules correspond to known lung-cancer biology?

We take the top-100 genes from every integration level, embed them in a STRING PPI network, and identify modules using the same three algorithms as in the course lab:
- **MCODE** (Bader & Hogue 2003) — densest local sub-graph in the seed-gene neighbourhood
- **DIAMOnD** (Ghiassian et al. 2015) — iterative seed-and-expand by hypergeometric connectivity
- **Clique Sum** (Tovar et al.) — clique-based aggregation; seeds assigned uniform p = 0.001

Pathway enrichment (KEGG 2021, MSigDB Hallmarks 2020, DisGeNET) is run on each MCODE module. §13.6 compares pathways across all levels.

| Level | Gene source |
|---|---|
| RNASeq | `stab_results` / RF top importance (§12.2) |
| DNAm | `shap_gene_dnam` — CpG→gene mapped (§11.1) |
| CNV | `stab_results` / RF top importance (§12.2) |
| RNASeq+DNAm | `pair_interp_models` gene features + CpG-mapped (§12.4a) |
| RNASeq+CNV | `pair_interp_models` gene features (§12.4a) |
| DNAm+CNV | `pair_interp_models` gene features + CpG-mapped (§12.4a) |
| Consensus | cross-layer rank aggregation `consensus` (§12.6) |
| DIABLO | `diablo_loadings` comp1 (only if §7 ran) |


```python
# ── 13.1  PPI network + module identification helpers ─────────────────────────
import sqlite3, math
import networkx as nx
from scipy.stats import hypergeom, combine_pvalues

_LAB_DATA = '../CompSys_2026_installaton/Data/'
PPI_PATH  = _LAB_DATA + 'PPI_network.txt'
CLIQUE_DB = _LAB_DATA + 'cliqueDB_700.sqlite'
PPI_SCORE = 700   # minimum STRING combined score

ppi13   = pd.read_csv(PPI_PATH, sep='\t', dtype=str)
G_ppi13 = nx.from_pandas_edgelist(
    ppi13[ppi13['combined_score'].astype(int) >= PPI_SCORE],
    source='entrez1', target='entrez2')
print(f'PPI (score >= {PPI_SCORE}): {G_ppi13.number_of_nodes():,} nodes, '
      f'{G_ppi13.number_of_edges():,} edges')


# ── MCODE (Bader & Hogue 2003) — identical to lab implementation ──────────────
def run_mcode13(G, seed_genes=None, vwp=0.5, haircut=False):
    if seed_genes is not None:
        nodes = [g for g in seed_genes if g in G]
        G_sub = G.subgraph(nodes).copy()
    else:
        G_sub = G.copy()
    if G_sub.number_of_nodes() == 0:
        return []
    weights = {}
    for v in G_sub.nodes():
        nbrs = list(G_sub.neighbors(v)) + [v]
        sg   = G_sub.subgraph(nbrs)
        k    = len(nbrs)
        d    = nx.density(sg) if k > 1 else 0.0
        weights[v] = k * d
    if not weights:
        return []
    seed        = max(weights, key=weights.get)
    seed_weight = weights[seed]
    if seed_weight == 0:
        return [seed]
    threshold = vwp * seed_weight
    cluster   = {seed}
    frontier  = set(G_sub.neighbors(seed))
    while frontier:
        new_frontier = set()
        for node in frontier:
            if node not in cluster and weights.get(node, 0) >= threshold:
                cluster.add(node)
                new_frontier.update(G_sub.neighbors(node))
        frontier = new_frontier - cluster
    if haircut:
        sg = G_sub.subgraph(cluster).copy()
        cluster -= {n for n in sg.nodes() if sg.degree(n) < 2}
    return list(cluster)


# ── DIAMOnD (Ghiassian et al. 2015) — identical to lab implementation ─────────
def run_diamond13(G, seed_genes, n_output_genes=200, seed_weight=10, include_seed=True):
    all_nodes = set(G.nodes())
    module    = set(g for g in seed_genes if g in all_nodes)
    N         = G.number_of_nodes()
    added     = []
    for _ in range(n_output_genes):
        candidates = set()
        for g in module:
            candidates.update(G.neighbors(g))
        candidates -= module
        if not candidates:
            break
        s = len(module)
        best_p, best_gene = 1.1, None
        for c in candidates:
            k = sum(1 for nb in G.neighbors(c) if nb in module)
            d = G.degree(c)
            p = hypergeom.sf(k - 1, N, s, d)
            if p < best_p:
                best_p, best_gene = p, c
        if best_gene is not None:
            module.add(best_gene)
            added.append(best_gene)
    return list(module) if include_seed else added


# ── Clique Sum (MODifieR) — identical to lab implementation ───────────────────
def run_clique_sum13(db_path, gene_pvalues, clique_significance=0.01,
                     deg_cutoff=0.05, min_clique_size=3):
    sig_genes = {g: p for g, p in gene_pvalues.items() if float(p) < float(deg_cutoff)}
    conn  = sqlite3.connect(db_path)
    cur   = conn.cursor()
    cur.execute("SELECT name FROM sqlite_master WHERE type='table' AND name LIKE 'clique%'")
    tables = sorted([r[0] for r in cur.fetchall()
                     if r[0].replace('clique', '').isdigit()],
                    key=lambda x: int(x.replace('clique', '')))
    module_genes = set()
    for table in tables:
        if int(table.replace('clique', '')) < min_clique_size:
            continue
        cur.execute(f'SELECT * FROM {table}')
        for row in cur.fetchall():
            genes  = [str(g) for g in row]
            in_sig = [g for g in genes if g in sig_genes]
            if not in_sig:
                continue
            if float(clique_significance) >= 1.0:
                module_genes.update(genes)
            else:
                pvals = [max(float(sig_genes[g]), 1e-300) for g in in_sig]
                cp    = pvals[0] if len(pvals) == 1 else combine_pvalues(pvals, method='fisher')[1]
                if cp < float(clique_significance):
                    module_genes.update(genes)
    conn.close()
    return list(module_genes)


# ── Enrichment wrapper (accepts gene symbols directly) ────────────────────────
def run_enrichr13(symbols, gene_sets, pval_cutoff=0.05):
    syms = [s for s in symbols if isinstance(s, str) and s and not s.isdigit()]
    if len(syms) < 5:
        return pd.DataFrame()
    try:
        enr = gp.enrichr(gene_list=syms, gene_sets=gene_sets, outdir=None, verbose=False)
        df  = enr.results[enr.results['Adjusted P-value'] < pval_cutoff]
        return df.sort_values('Adjusted P-value').reset_index(drop=True)
    except Exception as e:
        print(f'    Enrichr error: {e}')
        return pd.DataFrame()


print('Section 13 helpers defined.')
```

```python
# ── 13.2  Collect top-100 gene symbols per level + convert to Entrez IDs ───────
TOP_N_NET = 100   # top genes per integration level

try:
    _cpg_map = cpg_gene_map_std   # CpG -> standardised gene symbols (from §11.1)
except NameError:
    _cpg_map = {}
    print('cpg_gene_map_std not found — CpG probes in pairwise levels will be skipped.')


def _top_single(name, n=TOP_N_NET):
    # Top-n gene symbols from single-omics attribution (§12.2)
    r  = stab_results[name]
    fn = interp_data[name]['feat_names']
    m  = interp_models[name]
    return (list(fn[r['stable']])[:n] if len(r['stable']) >= 10
            else list(fn[np.argsort(m[IMP_KEYS[0]])[-n:][::-1]]))


def _top_pairwise(pair_name, n=TOP_N_NET):
    # Top-n gene symbols from pairwise attribution (§12.4a).
    # CpG probe IDs are mapped to genes via cpg_gene_map_std when available.
    fn  = pair_interp_data[pair_name]['feat_names']
    imp = pair_interp_models[pair_name][IMP_KEYS[0]]
    genes, seen = [], set()
    for f in fn[np.argsort(imp)[::-1]]:
        raw = f.split('_', 1)[1] if '_' in f else f
        if raw.startswith('cg'):
            for g in _cpg_map.get(raw, []):
                if g not in seen:
                    seen.add(g)
                    genes.append(g)
        else:
            if raw not in seen:
                seen.add(raw)
                genes.append(raw)
        if len(genes) >= n:
            break
    return genes[:n]


# Assemble per-level gene symbol lists
level_symbols13 = {
    'RNASeq':      _top_single('RNASeq'),
    'DNAm':        shap_gene_dnam.head(TOP_N_NET)['gene'].tolist(),
    'CNV':         _top_single('CNV'),
    'RNASeq+DNAm': _top_pairwise('RNASeq+DNAm'),
    'RNASeq+CNV':  _top_pairwise('RNASeq+CNV'),
    'DNAm+CNV':    _top_pairwise('DNAm+CNV'),
    'Consensus':   list(consensus.index[:TOP_N_NET]),
}

if DIABLO_AVAILABLE and diablo_loadings:
    diablo_syms, seen_d = [], set()
    for bl, dat in diablo_loadings.items():
        for g, _ in dat.get('comp1', [])[:50]:
            for gg in (_cpg_map.get(g, []) if g.startswith('cg') else [g]):
                if gg not in seen_d:
                    seen_d.add(gg)
                    diablo_syms.append(gg)
    level_symbols13['DIABLO'] = diablo_syms[:TOP_N_NET]

# Batch symbol -> Entrez conversion via mygene (single call for all levels)
_all_syms = sorted({g for gs in level_symbols13.values() for g in gs})
_mg       = mygene.MyGeneInfo()
_hits     = _mg.querymany(_all_syms, scopes='symbol', fields='entrezgene',
                           species='human', verbose=False)
sym2ent13 = {r['query']: str(int(r['entrezgene'])) for r in _hits
             if 'entrezgene' in r and not r.get('notfound')}
ent2sym13 = {v: k for k, v in sym2ent13.items()}   # Entrez -> symbol (reverse)

# Per-level Entrez lists — only genes present in PPI
level_entrez13 = {
    lv: [sym2ent13[g] for g in gs if g in sym2ent13 and sym2ent13[g] in G_ppi13]
    for lv, gs in level_symbols13.items()
}

print(f'{"Level":20s}  {"Symbols":>8}  {"in PPI":>8}')
print('-' * 42)
for lv, ids in level_entrez13.items():
    print(f'{lv:20s}  {len(level_symbols13[lv]):8d}  {len(ids):8d}')
```

```python
# ── 13.3  Module identification: MCODE + DIAMOnD + Clique Sum ─────────────────
# Seeds = top-100 important genes (Entrez IDs in PPI).
# Clique Sum uses uniform p = 0.001 for all seeds — all are equally 'significant'
# important features, analogous to DEG seeds used in the lab notebook.

DIAMOND_N = 200   # genes added in DIAMOnD expansion

module_results13 = {}   # {level: {'MCODE': [...], 'DIAMOnD': [...], 'CliqueSum': [...]}}

print(f'{"Level":20s}  {"Seeds":>6}  {"MCODE":>6}  {"DIAMOnD":>8}  {"CliqueSum":>10}')
print('-' * 60)

for level, eids in level_entrez13.items():
    if len(eids) < 5:
        print(f'{level:20s}  {len(eids):6d}  (too few seeds — skipped)')
        continue

    gene_pvalues = {e: 0.001 for e in eids}   # uniform significance for all seeds

    mcode   = run_mcode13(G_ppi13, seed_genes=eids, vwp=0.5, haircut=False)
    diamond = run_diamond13(G_ppi13, seed_genes=eids,
                            n_output_genes=DIAMOND_N, include_seed=True)
    clique  = run_clique_sum13(CLIQUE_DB, gene_pvalues=gene_pvalues,
                               clique_significance=0.01, deg_cutoff=0.05, min_clique_size=3)

    module_results13[level] = {'MCODE': mcode, 'DIAMOnD': diamond, 'CliqueSum': clique}
    print(f'{level:20s}  {len(eids):6d}  {len(mcode):6d}  {len(diamond):8d}  {len(clique):10d}')
```

### 13.7  Network Distance: Module Compactness and Cross-Omics Separation

Beyond identifying *which* genes form modules, we ask *how tight* each module
is, and *how close* different omics layers' modules are to each other in the PPI.

**Within-level compactness — d\u209b (Menche et al. 2015 *Science*)**  
The mean shortest path between all pairs of seed genes within a level.
A small d\u209b means the genes form a tight local neighbourhood — a hallmark of
a coherent disease module. Typical PPI average path length is ~3–4 hops.

**Cross-level network separation — s\u2090\u2099**  
For two gene sets A and B:

$$s_{AB} = d_{AB} - \frac{d_A + d_B}{2}$$

where $d_{AB}$ is the mean shortest path *between* the two sets and
$d_A$, $d_B$ are the within-set means. Interpretation:

| s\u2090\u2099 | Meaning |
|---|---|
| < 0 | Modules **overlap** — same PPI neighbourhood |
| \u2248 0 | Modules **touch** — adjacent neighbourhoods |
| > 0 | Modules **separated** — distinct network regions |

Both metrics use `G_ppi13` and `level_entrez13` defined in §13.1–13.2.
No data reload needed.

```python
# ── 13.7b  BFS from all seed genes → cache pairwise distances ─────────────────
# All unique seed genes across every level that are present in the PPI
_all_seeds = sorted({g for gs in level_entrez13.values() for g in gs if g in G_ppi13})
_seed_set  = set(_all_seeds)

print(f'Running BFS from {len(_all_seeds)} unique seed genes in PPI ...')
_dist_cache = {}
for _src in _all_seeds:
    lengths = nx.single_source_shortest_path_length(G_ppi13, _src)
    # Keep distances only to other seed genes (saves memory)
    _dist_cache[_src] = {t: d for t, d in lengths.items()
                         if t in _seed_set and t != _src}
print('BFS complete.')


def _d_within(nodes):
    # Mean shortest path within a gene set; returns (mean, n_reachable, n_total_pairs)
    valid = [n for n in nodes if n in _dist_cache]
    dists = []
    for i, u in enumerate(valid):
        for v in valid[i + 1:]:
            d = _dist_cache[u].get(v)
            if d is not None:
                dists.append(d)
    n_total = len(valid) * (len(valid) - 1) // 2
    return (np.mean(dists) if dists else np.nan), len(dists), n_total


def _d_between(nodes_a, nodes_b):
    # Mean shortest path from every gene in A to every gene in B
    set_b = set(nodes_b)
    dists = [d for u in nodes_a if u in _dist_cache
             for v, d in _dist_cache[u].items() if v in set_b]
    return np.mean(dists) if dists else np.nan


# ── Within-level compactness (d_S) ────────────────────────────────────────────
ds_results = {}
print(f'\n{"Level":20s}  {"d_S":>6}  {"Reachable pairs":>16}')
print('-' * 50)
for level, eids in level_entrez13.items():
    d_s, n_conn, n_total = _d_within(eids)
    ds_results[level] = d_s
    print(f'{level:20s}  {d_s:6.3f}  {n_conn:6d} / {n_total:6d}')

# ── Pairwise sAB matrix ────────────────────────────────────────────────────────
levels_ord = list(level_entrez13.keys())
n_lv       = len(levels_ord)
_sAB = np.full((n_lv, n_lv), np.nan)
_dAB = np.full((n_lv, n_lv), np.nan)

for i in range(n_lv):
    for j in range(i, n_lv):
        d_i  = ds_results[levels_ord[i]]
        d_j  = ds_results[levels_ord[j]]
        if i == j:
            _dAB[i, j] = d_i
        else:
            d_ij = _d_between(level_entrez13[levels_ord[i]],
                              level_entrez13[levels_ord[j]])
            _dAB[i, j] = _dAB[j, i] = d_ij
            s_ij = d_ij - (d_i + d_j) / 2
            _sAB[i, j] = _sAB[j, i] = s_ij

sAB_df = pd.DataFrame(_sAB, index=levels_ord, columns=levels_ord)
print('\nsAB matrix (negative = overlapping, positive = separated):')
display(sAB_df.round(3))
```

```python
# ── 13.7c  Visualise compactness and sAB ──────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(17, max(5, n_lv * 0.7 + 2)))

# ── Left: d_S bar chart ───────────────────────────────────────────────────────
ax = axes[0]
level_names = levels_ord
d_s_vals    = [ds_results[l] for l in level_names]
PPI_AVG     = 3.5   # typical mean shortest path in high-confidence STRING PPI
bar_colors  = ['#2ECC71' if (not np.isnan(v) and v < PPI_AVG) else '#E74C3C'
               for v in d_s_vals]
bars = ax.barh(level_names, d_s_vals, color=bar_colors, alpha=0.82, edgecolor='white')
ax.axvline(PPI_AVG, color='#7F8C8D', linestyle='--', linewidth=1.2,
           label=f'PPI avg ≈ {PPI_AVG}')
ax.set_xlabel('Mean shortest path within level (d_S)', fontsize=10)
ax.set_title('Module compactness per integration level\n'
             'green = tighter than PPI average, red = looser', fontsize=10)
ax.legend(fontsize=8)
ax.set_xlim(left=0)
for bar, val in zip(bars, d_s_vals):
    if not np.isnan(val):
        ax.text(bar.get_width() + 0.05, bar.get_y() + bar.get_height() / 2,
                f'{val:.2f}', va='center', fontsize=8)

# ── Right: sAB heatmap ────────────────────────────────────────────────────────
ax = axes[1]
# Mask diagonal (self-comparison is meaningless for sAB)
sAB_vis = sAB_df.copy()
np.fill_diagonal(sAB_vis.values, np.nan)
off_diag = _sAB[~np.isnan(_sAB)]
vmax = float(np.nanmax(np.abs(off_diag))) if len(off_diag) > 0 else 1.0
sns.heatmap(sAB_vis, cmap='RdBu_r', center=0, vmin=-vmax, vmax=vmax,
            annot=True, fmt='.2f', linewidths=0.4, linecolor='#DDDDDD',
            ax=ax, cbar_kws={'label': 's_AB'},
            xticklabels=True, yticklabels=True)
ax.set_xticklabels(ax.get_xticklabels(), rotation=35, ha='right', fontsize=8)
ax.set_yticklabels(ax.get_yticklabels(), rotation=0, fontsize=8)
ax.set_title('Network separation (s_AB) between integration levels\n'
             'blue (< 0) = overlapping modules, red (> 0) = separated', fontsize=10)

plt.suptitle('PPI network distances — compactness and cross-level separation',
             fontsize=12, y=1.01)
plt.tight_layout()
plt.show()
```
