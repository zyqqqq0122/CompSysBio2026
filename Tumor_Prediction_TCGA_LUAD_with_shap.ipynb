# Tumor Prediction — TCGA-LUAD Multi-Omics

This notebook trains and evaluates **tumour vs. normal classifiers** using three omics modalities
individually and in pairwise early-fusion combinations.

| Layer | Samples | Features |
|-------|---------|----------|
| RNASeq | 447 | 14 434 genes |
| DNAm | 420 | 384 629 CpG sites |
| CNV | 424 | 14 434 genes |

**Pipeline per experiment**
1. Variance filter — keep top *N* most variable features per modality
2. StandardScaler → fit classifier → 5-fold stratified cross-validation
3. Report AUC-ROC, Balanced Accuracy, F1 (macro), Accuracy

**Models**
- Logistic Regression (L2, PCA-compressed, class-weighted)
- Random Forest (200 trees, PCA-compressed, class-weighted)
- Linear SVM (Platt-calibrated probabilities, PCA-compressed, class-weighted)
- Elastic Net (L1+L2 logistic regression, raw features, class-weighted)
- XGBoost (gradient boosting, raw features)

```python
def harmonise_index(df):
    """
    Normalize TCGA sample IDs to 12-char patient IDs.
    TCGA barcodes appear with '.', '-', or '_' as separators depending on the
    source file; all three must be collapsed to '-' before truncating.
    Deduplicates so each patient contributes at most one row per modality.
    """
    df = df.copy()
    df.index = (df.index
                .str.replace('.', '-', regex=False)
                .str.replace('_', '-', regex=False)
                .str[:12])
    df = df.loc[~df.index.duplicated(keep='first')]
    return df

rnaseq = harmonise_index(rnaseq)
dnam   = harmonise_index(dnam)
cnv    = harmonise_index(cnv)

# Apply the same normalisation to metadata patient IDs before building the map.
# drop_duplicates handles the rare case where a patient appears more than once.
pid_norm  = (meta['patient_id']
             .str.replace('.', '-', regex=False)
             .str.replace('_', '-', regex=False)
             .str[:12])
label_map = (pd.Series(meta['label'].values, index=pid_norm)
             .loc[lambda s: ~s.index.duplicated(keep='first')])

# Coverage check
print(f'Unique patients in metadata: {len(label_map)}')
print(f'  label=0 (Normal): {(label_map == 0).sum()}')
print(f'  label=1 (Tumour): {(label_map == 1).sum()}\n')
for name, df in [('RNASeq', rnaseq), ('DNAm', dnam), ('CNV', cnv)]:
    n_matched = (~df.index.map(label_map).isna()).sum()
    print(f'{name:8s}  {df.shape[0]:4d} samples  →  {n_matched} matched in metadata')
```

## 2. Preprocessing & Feature Selection

The pipeline applies **two sequential dimensionality-reduction steps**:

1. **Variance filter** — retain the top *N* most variable features per modality.
   Fast unsupervised pre-filtering to a tractable range before PCA.
   For DNAm this is especially important: the 384 629 CpG sites are massively
   correlated because many belong to the same gene or regulatory region,
   so even the top-2000 subset will contain strong co-linear structure.

2. **PCA (Principal Component Analysis)** — applied *inside the sklearn pipeline*,
   fit only on training data in each CV fold to prevent leakage.
   PCA decorrelates the filtered features and compresses them into *K* orthogonal
   components that capture the dominant axes of variation.

**Pairwise early fusion** applies modality-specific scaling + PCA to each block
independently (`ColumnTransformer`), then concatenates the *K* components from
both modalities. This prevents one modality from dominating the joint feature space.

```python
def select_top_var(df, n):
    return df[df.var(axis=0).nlargest(n).index]


def prepare_single(df, label_map, n_features):
    """
    Variance filter → label lookup via patient ID → (X, y).
    Samples whose patient ID is absent from label_map are dropped.
    """
    df_sel = select_top_var(df, n_features)
    labels = df_sel.index.map(label_map)
    valid  = ~labels.isna()
    X = df_sel.loc[valid].values.astype(float)
    y = labels[valid].astype(int).values
    return X, y


def prepare_pair(df1, df2, label_map, n_features, prefix1, prefix2):
    """
    Concatenate top features from two modalities on their shared patient IDs.
    Label is looked up from metadata via the (already-normalised) 12-char patient ID.
    Samples without a metadata match are excluded.
    """
    sel1 = select_top_var(df1, n_features).add_prefix(prefix1 + '_')
    sel2 = select_top_var(df2, n_features).add_prefix(prefix2 + '_')
    shared = sel1.index.intersection(sel2.index)
    X      = np.hstack([sel1.loc[shared].values,
                        sel2.loc[shared].values]).astype(float)
    labels = pd.Series(shared.map(label_map), index=shared)
    valid  = ~labels.isna()
    X = X[valid.values]
    y = labels[valid].astype(int).values
    return X, y


print('Preprocessing helpers defined.')
```

## 3. Classifiers & Evaluation Framework

**PCA policy (applies to ALL models):**
- **RNASeq** and **DNAm**: always compressed via `StandardScaler → PCA(K)` before the classifier.
- **CNV**: always uses raw variance-filtered features (`StandardScaler` only — no PCA).

**Single-omics pipeline per fold:**
> Variance filter → StandardScaler [→ PCA(*K*) if RNA/DNAm] → Classifier

**Pairwise pipeline per fold:**
> [mod 1: Scaler → PCA(*K*) if RNA/DNAm] ⊕ [mod 2: Scaler → PCA(*K*) if RNA/DNAm] → Concat → Classifier
> Each modality block processed independently via `ColumnTransformer`.

All class-weighted models use `class_weight='balanced'` to handle tumour/normal imbalance.

| Model | Regularisation | PCA (RNA/DNAm) | PCA (CNV) |
|-------|---------------|:--------------:|:---------:|
| Logistic Regression | L2, C = 1.0 | ✓ | ✗ |
| Random Forest | 200 trees | ✓ | ✗ |
| Linear SVM | C = 1.0, Platt-scaled | ✓ | ✗ |
| Elastic Net | L1+L2, l1_ratio = 0.5, C = 1.0 | ✓ | ✗ |
| XGBoost | 100 trees, lr = 0.05, depth = 3 | ✓ | ✗ |

**Active models** are controlled by `MODELS_TO_RUN` in the config cell.

**Evaluation**: 5-fold stratified CV. All preprocessing steps are re-fit on training data each fold — no leakage.

```python
# PCA diagnostic: explained variance per modality
# Fitted on all labelled samples to guide the N_PCA choice.
# In the actual classification pipeline, PCA is re-fit on each training fold only.
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, (name, df, col) in zip(axes, [('RNASeq', rnaseq, PALETTE['RNASeq']),
                                        ('DNAm',   dnam,   PALETTE['DNAm']),
                                        ('CNV',    cnv,    PALETTE['CNV'])]):
    labels = df.index.map(label_map)
    df_sel = select_top_var(df, N_SINGLE)
    valid  = ~labels.isna()
    X_sub  = df_sel.loc[valid].values.astype(float)

    n_show = min(N_PCA * 2, 100)
    n_comp = min(n_show, X_sub.shape[0] - 1, X_sub.shape[1])
    pca_diag = PCA(n_components=n_comp, random_state=SEED)
    pca_diag.fit(StandardScaler().fit_transform(X_sub))

    evr    = pca_diag.explained_variance_ratio_ * 100
    cumvar = np.cumsum(evr)
    xs     = range(1, len(evr) + 1)

    ax.bar(xs, evr, color=col, alpha=0.6, label='Per-PC variance')
    ax2 = ax.twinx()
    ax2.plot(xs, cumvar, 'r-o', markersize=3, lw=1.5, label='Cumulative %')
    ax2.set_ylim(0, 105)
    ax2.set_ylabel('Cumulative variance (%)', fontsize=8)

    ax.axvline(N_PCA, ls='--', color='black', lw=1.5, label=f'N_PCA = {N_PCA}')
    ax2.axhline(95, ls=':', color='grey', lw=1, alpha=0.8)

    n_for_95   = int(np.argmax(cumvar >= 95)) + 1
    cum_at_cut = cumvar[min(N_PCA - 1, len(cumvar) - 1)]
    ax.text(0.97, 0.55,
            f'PCs for 95% var: {n_for_95}\nVar at PC {N_PCA}: {cum_at_cut:.1f}%',
            transform=ax.transAxes, ha='right', va='bottom', fontsize=8,
            bbox=dict(boxstyle='round', facecolor='white', alpha=0.85))

    ax.set_title(f'{name}  (top {N_SINGLE} features → PCA, n={valid.sum()})', fontsize=10)
    ax.set_xlabel('Principal Component')
    ax.set_ylabel('Variance explained (%)')
    lines1, labs1 = ax.get_legend_handles_labels()
    lines2, labs2 = ax2.get_legend_handles_labels()
    ax.legend(lines1 + lines2, labs1 + labs2, fontsize=7, loc='upper right')

plt.suptitle('PCA scree plots per modality\n'
             '(diagnostic fit on all labelled samples; actual CV fits only on training folds)',
             fontsize=12)
plt.tight_layout()
plt.show()
```

```python
def make_single_models(use_pca=True):
    """
    Build all five classifiers with a uniform PCA policy.
    use_pca=True  → StandardScaler → PCA(N_PCA) → Classifier  (RNA and DNAm)
    use_pca=False → StandardScaler → Classifier               (CNV)
    Only models listed in MODELS_TO_RUN are returned.
    """
    def build(clf):
        steps = [('scaler', StandardScaler())]
        if use_pca:
            steps.append(('pca', PCA(n_components=N_PCA, random_state=SEED)))
        steps.append(('clf', clf))
        return Pipeline(steps)

    all_models = {
        'Logistic Regression': build(LogisticRegression(
            C=1.0, max_iter=2000, class_weight='balanced',
            solver='saga', random_state=SEED)),
        'Random Forest': build(RandomForestClassifier(
            n_estimators=200, max_features='sqrt',
            class_weight='balanced', random_state=SEED, n_jobs=-1)),
        'Linear SVM': build(SVC(
            kernel='linear', C=1.0, probability=True,
            class_weight='balanced', random_state=SEED)),
        'Elastic Net': build(LogisticRegression(
            penalty='elasticnet', solver='saga', l1_ratio=0.5, C=1.0,
            max_iter=5000, class_weight='balanced', random_state=SEED)),
        'XGBoost': build(XGBClassifier(
            n_estimators=100, learning_rate=0.05, max_depth=3,
            eval_metric='logloss', random_state=SEED)),
    }
    return {k: v for k, v in all_models.items() if k in MODELS_TO_RUN}


def make_pair_models(n_per_modality=N_PAIR, pca_mod1=True, pca_mod2=True):
    """
    All classifiers use a per-modality ColumnTransformer so PCA is applied
    independently to each block.
    pca_mod1 / pca_mod2 control whether each modality block is PCA-compressed.
    Only models listed in MODELS_TO_RUN are returned.
    """
    def mod_pipe(use_pca):
        steps = [('scaler', StandardScaler())]
        if use_pca:
            steps.append(('pca', PCA(n_components=N_PCA, random_state=SEED)))
        return Pipeline(steps)

    transform = ColumnTransformer([
        ('mod1', mod_pipe(pca_mod1), slice(0, n_per_modality)),
        ('mod2', mod_pipe(pca_mod2), slice(n_per_modality, 2 * n_per_modality)),
    ])

    classifiers = {
        'Logistic Regression': LogisticRegression(
            C=1.0, max_iter=2000, class_weight='balanced',
            solver='saga', random_state=SEED),
        'Random Forest': RandomForestClassifier(
            n_estimators=200, max_features='sqrt',
            class_weight='balanced', random_state=SEED, n_jobs=-1),
        'Linear SVM': SVC(
            kernel='linear', C=1.0, probability=True,
            class_weight='balanced', random_state=SEED),
        'Elastic Net': LogisticRegression(
            penalty='elasticnet', solver='saga', l1_ratio=0.5, C=1.0,
            max_iter=5000, class_weight='balanced', random_state=SEED),
        'XGBoost': XGBClassifier(
            n_estimators=100, learning_rate=0.05, max_depth=3,
            eval_metric='logloss', random_state=SEED),
    }

    return {
        name: Pipeline([('transform', clone(transform)), ('clf', clf)])
        for name, clf in classifiers.items()
        if name in MODELS_TO_RUN
    }


# Derived from config — used by downstream plotting cells
models_list = list(MODELS_TO_RUN)


def evaluate_dataset(X, y, model_factory, n_splits=CV_FOLDS):
    """
    Stratified K-fold CV for all models from model_factory().
    Automatically reduces n_splits if the minority class is too small.
    All preprocessing steps are re-fit on each training fold only — no data leakage.
    """
    counts = np.bincount(y)
    min_class = counts.min()

    effective_splits = min(n_splits, min_class)
    if effective_splits < 2:
        print(f'    WARNING: only {min_class} minority-class sample(s) — '
              f'cannot run CV, skipping this dataset.')
        nan_metrics = {k: np.nan for k in
                       ['AUC-ROC', 'AUC std', 'Balanced Acc', 'F1 (macro)', 'Accuracy']}
        dummy = {'pool_y_true': y,
                 'pool_y_score': np.full_like(y, 0.5, dtype=float)}
        return {m: {**nan_metrics, **dummy} for m in model_factory().keys()}

    if effective_splits < n_splits:
        print(f'    Note: using {effective_splits}-fold CV '
              f'(minority class has only {min_class} samples)')

    skf = StratifiedKFold(n_splits=effective_splits, shuffle=True, random_state=SEED)
    results = {}

    for mname, pipeline in model_factory().items():
        fold_aucs, fold_baccs, fold_f1s, fold_accs = [], [], [], []
        pool_y_true, pool_y_score = [], []

        for train_idx, test_idx in skf.split(X, y):
            m = clone(pipeline)
            m.fit(X[train_idx], y[train_idx])
            y_pred  = m.predict(X[test_idx])
            y_score = m.predict_proba(X[test_idx])[:, 1]

            fold_aucs.append(roc_auc_score(y[test_idx], y_score))
            fold_baccs.append(balanced_accuracy_score(y[test_idx], y_pred))
            fold_f1s.append(f1_score(y[test_idx], y_pred, average='macro'))
            fold_accs.append(accuracy_score(y[test_idx], y_pred))
            pool_y_true.extend(y[test_idx].tolist())
            pool_y_score.extend(y_score.tolist())

        results[mname] = {
            'AUC-ROC':      round(np.mean(fold_aucs),  4),
            'AUC std':      round(np.std(fold_aucs),   4),
            'Balanced Acc': round(np.mean(fold_baccs), 4),
            'F1 (macro)':   round(np.mean(fold_f1s),   4),
            'Accuracy':     round(np.mean(fold_accs),  4),
            'pool_y_true':  np.array(pool_y_true),
            'pool_y_score': np.array(pool_y_score),
        }

    return results


print('Model factories and evaluation function defined.')
print(f'  Active models: {models_list}')
print(f'  PCA (RNA/DNAm): StandardScaler → PCA({N_PCA}) → Classifier')
print(f'  No PCA (CNV):   StandardScaler → Classifier')
print(f'  Pairwise:       ColumnTransformer([mod1 pipe, mod2 pipe]) → Concat → Classifier')
```

## 4. Single-Omics Classification

Each modality is evaluated independently.
Pipeline per fold: variance filter (top **2 000** features) → StandardScaler → PCA (**50** components) → Classifier.

```python
print(f'Feature selection: top {N_SINGLE} per modality → PCA({N_PCA})\n')

single_datasets = {
    'RNASeq': prepare_single(rnaseq, label_map, N_SINGLE),
    'DNAm':   prepare_single(dnam,   label_map, N_SINGLE),
    'CNV':    prepare_single(cnv,    label_map, N_SINGLE),
}

print('Dataset overview:')
print(f'{"Modality":<10} {"Samples":>8} {"Features":>10} {"Normal":>8} {"Tumour":>8}')
print('-' * 50)
for name, (X, y) in single_datasets.items():
    counts = np.bincount(y)
    print(f'{name:<10} {len(y):>8} {X.shape[1]:>10} {counts[0]:>8} {counts[1]:>8}')
```

```python
# CNV always skips PCA; RNA and DNAm always use PCA
SINGLE_PCA = {'RNASeq': True, 'DNAm': True, 'CNV': False}

print('Running single-omics cross-validation ...')
print(f'  PCA policy: {SINGLE_PCA}\n')

single_results = {}
for name, (X, y) in single_datasets.items():
    use_pca = SINGLE_PCA[name]
    print(f'  {name} (PCA={use_pca}) ...', end=' ', flush=True)
    single_results[name] = evaluate_dataset(
        X, y, lambda up=use_pca: make_single_models(use_pca=up))
    best_auc = max(v['AUC-ROC'] for v in single_results[name].values())
    print(f'best AUC-ROC = {best_auc:.3f}')

print('\nDone.')
```

```python
rows = []
for dataset, model_res in single_results.items():
    for model, metrics in model_res.items():
        rows.append({'Dataset': dataset, 'Model': model,
                     **{k: metrics[k] for k in METRIC_COLS}})

single_summary = pd.DataFrame(rows)
print('Single-omics results (mean over 5 folds):')
display(single_summary.set_index(['Dataset', 'Model']))
```

## 5. Pairwise Omics Combinations — Early Fusion

For each pair, the top **1 000** features from each modality are selected independently.
Each block is then scaled and projected to **50** PCA components via a `ColumnTransformer`,
giving **100** decorrelated features total before the classifier.
Only samples present in **both** modalities are used.

```python
print(f'Feature selection: top {N_PAIR} per modality → per-modality PCA({N_PCA}) → {2*N_PCA} dims\n')

pair_datasets = {
    'RNASeq+DNAm': prepare_pair(rnaseq, dnam, label_map, N_PAIR, 'RNA', 'DNAm'),
    'RNASeq+CNV':  prepare_pair(rnaseq, cnv,  label_map, N_PAIR, 'RNA', 'CNV'),
    'DNAm+CNV':    prepare_pair(dnam,   cnv,  label_map, N_PAIR, 'DNAm', 'CNV'),
}

print('Dataset overview:')
print(f'{"Combination":<16} {"Samples":>8} {"Features":>10} {"Normal":>8} {"Tumour":>8}')
print('-' * 56)
for name, (X, y) in pair_datasets.items():
    counts = np.bincount(y)
    print(f'{name:<16} {len(y):>8} {X.shape[1]:>10} {counts[0]:>8} {counts[1]:>8}')
```

```python
# PCA config per pair: (pca_mod1, pca_mod2)
# RNA and DNAm blocks use PCA; CNV block does not.
PAIR_PCA_CONFIG = {
    'RNASeq+DNAm': (True,  True),
    'RNASeq+CNV':  (True,  False),
    'DNAm+CNV':    (True,  False),
}

print('Running pairwise cross-validation ...')
print(f'  PCA config: {PAIR_PCA_CONFIG}\n')

pair_results = {}
for name, (X, y) in pair_datasets.items():
    pca1, pca2 = PAIR_PCA_CONFIG[name]
    print(f'  {name} (PCA mod1={pca1}, mod2={pca2}) ...', end=' ', flush=True)
    pair_results[name] = evaluate_dataset(
        X, y,
        lambda p1=pca1, p2=pca2: make_pair_models(N_PAIR, pca_mod1=p1, pca_mod2=p2))
    best_auc = max(v['AUC-ROC'] for v in pair_results[name].values())
    print(f'best AUC-ROC = {best_auc:.3f}')

print('\nDone.')
```

## 8. Biological Interpretation — SHAP Analysis (XGBoost)

**Why SHAP?**  
SHAP (SHapley Additive exPlanations) assigns each feature a contribution score per prediction.

**PCA backprojection for RNA and DNAm:**

Because XGBoost operates on PCA components, SHAP values directly measure *component importance*.
To recover *gene-level importance*, we project back through the PCA loading matrix:

$$\text{importance}_j = \sum_{k=1}^{K} \overline{|\text{SHAP}|}_k \cdot |W_{kj}|$$

where $\overline{|\text{SHAP}|}_k$ is the mean absolute SHAP value for PC $k$ and $W_{kj}$ is the loading
of gene $j$ on PC $k$. This is a single matrix multiply — computationally trivial.

**Summary of interpretability per modality:**
| Modality | SHAP features | Backprojection | Result |
|---------|--------------|---------------|--------|
| RNASeq | PCA components | ✓ direct | Named genes |
| DNAm | PCA components | ✓ direct | CpG probe IDs (need manifest → genes) |
| CNV | Raw genes (no PCA) | not needed | Named genes |

**Sections:**
- **8.1** — RNASeq: SHAP on components + backprojection to genes
- **8.2** — DNAm: SHAP on components + backprojection to CpG probes
- **8.3** — CNV: SHAP directly on gene-level features
- **8.4** — Cross-fold Jaccard stability (CNV)
- **Section 9** — Pairwise fusion SHAP with backprojection
- **Section 10** — KEGG/GO enrichment (RNASeq and CNV genes)

> **Requires:** `pip install shap`

```python
import shap

# Helper: project component-level SHAP back to original feature space
def shap_backproject(mean_abs_shap_per_component, pca):
    """
    mean_abs_shap_per_component : array of shape (n_components,)
    pca                         : fitted sklearn PCA object
    Returns                     : array of shape (n_features,) — one importance score per input feature
    """
    return np.abs(pca.components_).T @ mean_abs_shap_per_component


# ── 8.1 RNASeq ────────────────────────────────────────────────────────────────
X_rna_full, y_rna_full = prepare_single(rnaseq, label_map, N_SINGLE)
feature_names_rna = select_top_var(rnaseq, N_SINGLE).columns.tolist()

xgb_rna = clone(make_single_models(use_pca=True)['XGBoost'])
xgb_rna.fit(X_rna_full, y_rna_full)

X_rna_pca = xgb_rna[:-1].transform(X_rna_full)
feature_names_rna_pca = [f'PC{i}' for i in range(X_rna_pca.shape[1])]

explainer_rna = shap.TreeExplainer(xgb_rna.named_steps['clf'])
shap_values_rna = explainer_rna.shap_values(X_rna_pca)
mean_shap_rna = np.abs(shap_values_rna).mean(axis=0)

# --- Component-level plot ---
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

plt.sca(axes[0])
shap.summary_plot(shap_values_rna, X_rna_pca,
                  feature_names=feature_names_rna_pca, max_display=15, show=False)
axes[0].set_title('RNASeq — SHAP on PCA components')

# --- Backprojection to gene space ---
gene_imp_rna = shap_backproject(mean_shap_rna, xgb_rna.named_steps['pca'])
shap_gene_rna = (pd.DataFrame({'gene': feature_names_rna, 'importance': gene_imp_rna})
                   .sort_values('importance', ascending=False)
                   .reset_index(drop=True))

top_genes_plot = shap_gene_rna.head(20)
axes[1].barh(top_genes_plot['gene'][::-1], top_genes_plot['importance'][::-1],
             color=PALETTE['RNASeq'], alpha=0.85)
axes[1].set_xlabel('SHAP-weighted loading importance')
axes[1].set_title('RNASeq — top 20 genes (SHAP backprojected through PCA)')
axes[1].tick_params(axis='y', labelsize=8)

plt.suptitle('RNASeq XGBoost SHAP — component view and gene-level backprojection', fontsize=12)
plt.tight_layout()
plt.show()

print('Top 20 RNASeq genes by SHAP-PCA backprojection:')
display(shap_gene_rna.head(20))
```

```python
# 8.4 Cross-fold SHAP stability for CNV (gene-level, interpretable)
# RNA/DNAm use PCA components — Jaccard on component indices is less meaningful biologically.
top_n = 30

def compute_fold_jaccard(X_full, y_full, use_pca, top_n=30):
    fold_top = []
    skf_shap = StratifiedKFold(n_splits=CV_FOLDS, shuffle=True, random_state=SEED)
    for train_idx, test_idx in skf_shap.split(X_full, y_full):
        pipe = clone(make_single_models(use_pca=use_pca)['XGBoost'])
        pipe.fit(X_full[train_idx], y_full[train_idx])
        X_test_transformed = pipe[:-1].transform(X_full[test_idx])
        exp = shap.TreeExplainer(pipe.named_steps['clf'])
        sv  = exp.shap_values(X_test_transformed)
        top_idx = set(np.argsort(np.abs(sv).mean(axis=0))[-top_n:])
        fold_top.append(top_idx)
    return fold_top

def plot_jaccard(fold_top, title, ax):
    n   = len(fold_top)
    jac = np.zeros((n, n))
    for i in range(n):
        for j in range(n):
            inter = len(fold_top[i] & fold_top[j])
            union = len(fold_top[i] | fold_top[j])
            jac[i, j] = inter / union if union > 0 else 1.0
    sns.heatmap(jac, annot=True, fmt='.2f', cmap='Blues', vmin=0, vmax=1,
                xticklabels=[f'Fold {i+1}' for i in range(n)],
                yticklabels=[f'Fold {i+1}' for i in range(n)], ax=ax)
    ax.set_title(title)
    return jac[np.triu_indices(n, k=1)].mean()

fig, ax = plt.subplots(1, 1, figsize=(6, 5))

print('Computing cross-fold Jaccard for CNV (gene-level SHAP) ...')
fold_top_cnv = compute_fold_jaccard(X_cnv_full, y_cnv_full, use_pca=False, top_n=top_n)
jac_cnv = plot_jaccard(fold_top_cnv,
    f'Jaccard similarity — top-{top_n} SHAP genes\n(CNV, XGBoost, no PCA)', ax)

plt.suptitle(f'Cross-fold SHAP feature stability — top-{top_n} CNV genes, 5-fold CV', fontsize=13)
plt.tight_layout()
plt.show()

print(f'Mean cross-fold Jaccard — CNV: {jac_cnv:.3f}')
print('(> 0.5 = good stability; > 0.7 = strong stability)')
print('\nNote: Jaccard stability for RNASeq/DNAm is not shown — SHAP values')
print('index PCA components, so fold consistency of component indices is less')
```

## 9. SHAP Analysis — Pairwise Fusion XGBoost

Each pairwise XGBoost uses a `ColumnTransformer` that applies the same PCA policy as in CV:
- **RNASeq+DNAm**: both blocks PCA-compressed → SHAP maps to PCA components.
- **RNASeq+CNV**: RNA block PCA-compressed, CNV block raw → CNV side maps to gene names.
- **DNAm+CNV**: DNAm block PCA-compressed, CNV block raw → CNV side maps to gene names.

For combinations with a raw CNV block, per-modality SHAP contributions are computed to
quantify which data layer dominates the fusion model.

```python
# 9.1 RNASeq+DNAm — both blocks PCA-compressed; backproject both to original features
X_rna_dnam, y_rna_dnam = pair_datasets['RNASeq+DNAm']
feature_names_rna_pair  = select_top_var(rnaseq, N_PAIR).columns.tolist()
feature_names_dnam_pair = select_top_var(dnam,   N_PAIR).columns.tolist()

xgb_rna_dnam = clone(make_pair_models(N_PAIR, pca_mod1=True, pca_mod2=True)['XGBoost'])
xgb_rna_dnam.fit(X_rna_dnam, y_rna_dnam)

X_rna_dnam_tr = xgb_rna_dnam.named_steps['transform'].transform(X_rna_dnam)
feature_names_rna_dnam_tr = (
    [f'RNA_PC{i}' for i in range(N_PCA)] + [f'DNAm_PC{i}' for i in range(N_PCA)]
)

explainer_rna_dnam = shap.TreeExplainer(xgb_rna_dnam.named_steps['clf'])
shap_values_rna_dnam = explainer_rna_dnam.shap_values(X_rna_dnam_tr)
mean_shap_rna_dnam = np.abs(shap_values_rna_dnam).mean(axis=0)

# Split SHAP by block
mean_shap_rna_block  = mean_shap_rna_dnam[:N_PCA]
mean_shap_dnam_block = mean_shap_rna_dnam[N_PCA:]

# Retrieve per-block PCA objects from ColumnTransformer
pca_rna_block  = xgb_rna_dnam.named_steps['transform'].transformers_[0][1].named_steps['pca']
pca_dnam_block = xgb_rna_dnam.named_steps['transform'].transformers_[1][1].named_steps['pca']

gene_imp_rna_pair = shap_backproject(mean_shap_rna_block,  pca_rna_block)
cpg_imp_dnam_pair = shap_backproject(mean_shap_dnam_block, pca_dnam_block)

shap_gene_rna_dnam = pd.DataFrame({
    'feature':     feature_names_rna_pair + feature_names_dnam_pair,
    'modality':    ['RNASeq'] * N_PAIR    + ['DNAm']   * N_PAIR,
    'importance':  np.concatenate([gene_imp_rna_pair, cpg_imp_dnam_pair]),
}).sort_values('importance', ascending=False).reset_index(drop=True)

# --- Plots ---
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Component-level summary
plt.sca(axes[0])
shap.summary_plot(shap_values_rna_dnam, X_rna_dnam_tr,
                  feature_names=feature_names_rna_dnam_tr, max_display=15, show=False)
axes[0].set_title('RNASeq+DNAm — SHAP on PCA components')

# Backprojected top features, coloured by modality
top20 = shap_gene_rna_dnam.head(20)
colors = [PALETTE['RNASeq'] if m == 'RNASeq' else PALETTE['DNAm'] for m in top20['modality'][::-1]]
axes[1].barh(top20['feature'][::-1], top20['importance'][::-1], color=colors, alpha=0.85)
axes[1].set_xlabel('SHAP-weighted loading importance')
axes[1].set_title('RNASeq+DNAm — top 20 features (backprojected)')
axes[1].tick_params(axis='y', labelsize=8)
from matplotlib.patches import Patch
axes[1].legend(handles=[Patch(color=PALETTE['RNASeq'], label='RNASeq'),
                         Patch(color=PALETTE['DNAm'],   label='DNAm')], fontsize=8)

plt.suptitle('RNASeq+DNAm XGBoost SHAP — component view and feature backprojection', fontsize=12)
plt.tight_layout()
plt.show()

print('Top 20 backprojected features (RNASeq+DNAm):')
display(shap_gene_rna_dnam.head(20))
print('\nTotal backprojected importance by modality:')
print(shap_gene_rna_dnam.groupby('modality')['importance'].sum().to_string())

# Store top RNA genes for enrichment in section 10
top_genes_rna_in_rna_dnam = shap_gene_rna_dnam[shap_gene_rna_dnam['modality'] == 'RNASeq'].head(50)['feature'].tolist()
```

## 10. KEGG / GO Pathway Enrichment

Top SHAP genes from gene-annotated modalities (RNASeq, CNV) are tested for pathway enrichment
using the Enrichr API (requires internet access).

**Note on DNAm:** Features are CpG probe IDs (e.g. `cg21870274`), not gene names. Pathway enrichment
requires mapping probes to genes via the Illumina 450K/EPIC manifest — not performed here.

> **Requires:** `pip install gseapy`

**Interpretation guide:**
- Low adj. p-value in oncogenic pathways (*p53*, *EGFR*, *KRAS*) → model captures known LUAD biology.
- Cell-cycle / DNA repair enrichment from CNV features → copy-number instability driving tumour identity.
- Overlap between RNASeq and CNV top SHAP genes → co-regulated loci where both expression and copy number are disrupted.

```python
import gseapy as gp

def run_enrichr(gene_list, label):
    """Run Enrichr on a gene list and return significant results."""
    print(f'Running enrichment on {len(gene_list)} genes ({label}) ...')
    enr = gp.enrichr(
        gene_list=gene_list,
        gene_sets=['KEGG_2021_Human', 'GO_Biological_Process_2021'],
        organism='Human', outdir=None, verbose=False
    )
    sig = (enr.results[enr.results['Adjusted P-value'] < 0.05]
               .sort_values('Adjusted P-value'))
    print(f'  Significant pathways (adj. p < 0.05): {len(sig)}')
    return sig


# ── RNASeq top-50 genes (from SHAP backprojection) ───────────────────────────
top_genes_rna_single = shap_gene_rna.head(50)['gene'].tolist()
sig_rna = run_enrichr(top_genes_rna_single, 'RNASeq single-omics (backprojected)')
display(sig_rna[['Gene_set', 'Term', 'Overlap', 'P-value', 'Adjusted P-value', 'Genes']].head(20))

# ── CNV top-50 genes (direct SHAP, no PCA) ────────────────────────────────────
top_genes_cnv_single = shap_df_cnv.head(50)['gene'].tolist()
sig_cnv = run_enrichr(top_genes_cnv_single, 'CNV single-omics (direct SHAP)')
display(sig_cnv[['Gene_set', 'Term', 'Overlap', 'P-value', 'Adjusted P-value', 'Genes']].head(20))
```
